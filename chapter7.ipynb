## 7.1 Diagonalization of Symmetric Matrices  对称矩阵的对角化

**对称矩阵的定义**：若 $A^T=A$，则称 $A$ 为对称矩阵。对称矩阵必为方阵，且关于主对角线对称：

$$
a_{ij}=a_{ji}.
$$

例如

$$
\begin{bmatrix}
4&3\\
3&-8
\end{bmatrix},
\qquad
\begin{bmatrix}
-2&4&5\\
4&-2&3\\
5&3&-2
\end{bmatrix}
$$

都是对称矩阵。对称矩阵的主对角线元素任意，非对角线元素成对出现。常见的对称矩阵包括 $A^TA$、$AA^T$、$B^TB$、$BB^T$。


**定理 1（不同特征值的特征向量正交）**：若 $A$ 对称，则属于不同特征值的特征向量正交。

证明：设

$$
A\mathbf v_1=\lambda_1\mathbf v_1,\qquad
A\mathbf v_2=\lambda_2\mathbf v_2,\qquad
\lambda_1\neq\lambda_2.
$$

计算

$$
\lambda_1\mathbf v_1\cdot\mathbf v_2
=(A\mathbf v_1)^T\mathbf v_2
=\mathbf v_1^TA^T\mathbf v_2
=\mathbf v_1^TA\mathbf v_2
=\lambda_2\mathbf v_1\cdot\mathbf v_2.
$$

于是

$$
(\lambda_1-\lambda_2)\mathbf v_1\cdot\mathbf v_2=0.
$$

因为 $\lambda_1\neq\lambda_2$，所以 $\mathbf v_1\cdot\mathbf v_2=0$。


**正交矩阵与正交对角化**：正交矩阵满足 $P^TP=I$，等价于 $P^{-1}=P^T$。正交矩阵的列构成 $\mathbb R^n$ 的标准正交基。

若存在正交矩阵 $P$ 和对角矩阵 $D$，使得

$$
A=PDP^T=PDP^{-1},
$$

则称 $A$ 可正交对角化。

若 $A=PDP^T$，则

$$
A^T=(PDP^T)^T=PD^TP^T=PDP^T=A.
$$

所以**可正交对角化的矩阵必为对称矩阵**。反过来也成立，这就是定理 2。


**定理 2（对称 $\iff$ 可正交对角化）**：$n\times n$ 矩阵 $A$ 可正交对角化，当且仅当 $A$ 是对称矩阵。


**定理 3  谱定理 Spectral Theorem**

设 $A$ 为 $n\times n$ 实对称矩阵，则：

a. $A$ 有 $n$ 个实特征值（计重数）；  
b. 每个特征值的几何重数等于代数重数；  
c. 不同特征值对应的特征空间互相正交；  
d. $A$ 可正交对角化。

这说明对称矩阵不会出现“特征值不够”“特征向量不够”“特征向量不正交”的问题。

**谱分解 Spectral Decomposition**

若

$$
A=PDP^T,
$$

其中 $P=[\mathbf u_1\ \cdots\ \mathbf u_n]$，$D=\operatorname{diag}(\lambda_1,\dots,\lambda_n)$，则

$$
A=\lambda_1\mathbf u_1\mathbf u_1^T+\cdots+\lambda_n\mathbf u_n\mathbf u_n^T.
$$

这称为 $A$ 的**谱分解**。每一项 $\mathbf u_i\mathbf u_i^T$ 是到 $\operatorname{Span}\{\mathbf u_i\}$ 的正交投影矩阵。因此谱分解把 $A$ 拆成“沿各个正交主方向伸缩”的叠加。



谱分解里每一项 $\mathbf u_i\mathbf u_i^T$ 都是到 $\operatorname{Span}\{\mathbf u_i\}$ 的正交投影矩阵，秩为 1：它的每一列都是 $\mathbf u_i$ 的倍数，所有列都落在 $\mathbf u_i$ 这一条直线上。

由 $\mathbf y=P^T\mathbf x$，分量 $\mathbf u_i^T\mathbf x$ 是 $\mathbf x$ 沿第 $i$ 条主轴的投影长度。所以 $A\mathbf x=PD\mathbf y$ 分三步：投影到各主轴，沿主轴分别乘 $\lambda_i$，再把这些分量加回去。整个过程只有伸缩，没有旋转和扭曲。




**数值注意**：现代算法通常用一系列**正交相似变换**把对称矩阵逐步变成近似对角矩阵：

$$
A\longmapsto Q_1^TAQ_1\longmapsto Q_2^TQ_1^TAQ_1Q_2\longmapsto\cdots.
$$

使用正交矩阵可以避免误差累积，因此对称矩阵的特征值/特征向量计算通常非常稳定。

**False。**

**解：**

$$
(A\mathbf x)\cdot\mathbf y
=(A\mathbf x)^T\mathbf y
=\mathbf x^TA^T\mathbf y
=\mathbf x^TA\mathbf y
=\mathbf x\cdot(A\mathbf y).
$$





### 练习 7.1 36

**36. 若 $A,B$ 都可正交对角化，且 $AB=BA$，证明 $AB$ 也可正交对角化。**

**证明**：

因为 $A,B$ 可正交对角化，所以它们都是对称矩阵：

$$
A^T=A,\qquad B^T=B.
$$

又 $AB=BA$，于是

$$
(AB)^T=B^TA^T=BA=AB.
$$

所以 $AB$ 对称。由定理 2，$AB$ 可正交对角化。




### 练习 7.1 37

**37. $A=PDP^{-1}$，$P$ 正交，$D$ 对角。若 $\lambda$ 是 $A$ 的特征值，重数为 $k$，解释 $\lambda$ 的特征空间维数为 $k$。**

**解：**

因为 $P$ 正交，$A=PDP^T$，所以 $A$ 与 $D$ 相似。$\lambda$ 在 $D$ 的对角线上出现 $k$ 次。对角矩阵 $D$ 关于 $\lambda$ 的特征空间维数正好是 $k$（对应 $k$ 个标准基向量）。相似变换保持特征空间维数，因此

$$
\dim\operatorname{Nul}(A-\lambda I)=k.
$$



### 练习 7.1 38

**38. 设 $A=PRP^{-1}$，$P$ 正交，$R$ 上三角。若 $A$ 对称，证明 $R$ 对称，从而是对角矩阵。**

**证明：**

$$
R=P^TAP.
$$

于是

$$
R^T=(P^TAP)^T=P^TA^TP=P^TAP=R.
$$

所以 $R$ 对称。又 $R$ 是上三角矩阵，故 $R$ 必须是对角矩阵。

### 练习 7.1 42

设 $B$ 是一个 $n \times n$ 的对称矩阵，且满足 $B^2 = B$。任何这样的矩阵被称为**投影矩阵**（或**正交投影矩阵**）。给定 $\mathbb{R}^n$ 中的任意 $\mathbf{y}$，令 $\hat{\mathbf{y}} = B\mathbf{y}$ 且 $\mathbf{z} = \mathbf{y} - \hat{\mathbf{y}}$。

**a.** 证明 $\mathbf{z}$ 与 $\hat{\mathbf{y}}$ 正交。

**b.** 设 $W$ 是 $B$ 的列空间。证明 $\mathbf{y}$ 可以表示为 $W$ 中的一个向量与 $W^\perp$ 中的一个向量之和。为什么这能证明 $B\mathbf{y}$ 是 $\mathbf{y}$ 在 $B$ 的列空间上的正交投影？

**证明：**


**(a)**

$$
\mathbf z\cdot\hat{\mathbf y}
=(\mathbf y-B\mathbf y)^TB\mathbf y
=\mathbf y^TB\mathbf y-\mathbf y^TB^TB\mathbf y
=\mathbf y^TB\mathbf y-\mathbf y^TB^2\mathbf y=0.
$$

所以 $\mathbf z\perp\hat{\mathbf y}$。

**(b)** 令 $W=\operatorname{Col}B$。显然 $\hat{\mathbf y}=B\mathbf y\in W$。对任意 $B\mathbf v\in W$，

$$
\mathbf z\cdot B\mathbf v
=(\mathbf y-B\mathbf y)^TB\mathbf v
=\mathbf y^TB\mathbf v-\mathbf y^TB^2\mathbf v=0.
$$

所以 $\mathbf z\in W^\perp$。因此

$$
\mathbf y=\hat{\mathbf y}+\mathbf z,\qquad
\hat{\mathbf y}\in W,\quad \mathbf z\in W^\perp.
$$

由正交分解唯一性，$\hat{\mathbf y}=B\mathbf y$ 是 $\mathbf y$ 到 $\operatorname{Col}B$ 的正交投影。



### 练习 7.1 43

**43.** 正交对角化下面的矩阵。为了熟悉本节的方法，只用程序求特征值，不直接用程序的特征向量命令；再对每个特征值 $\lambda$ 求 $\operatorname{Nul}(A-\lambda I)$ 的一组标准正交基，做法同例 2、例 3。

$$
\begin{bmatrix}
6 & 2 & 9 & -6 \\
2 & 6 & -6 & 9 \\
9 & -6 & 6 & 2 \\
-6 & 9 & 2 & 6
\end{bmatrix}
$$


```python
import sympy as sp

A = sp.Matrix([
    [6, 2, 9, -6],
    [2, 6, -6, 9],
    [9, -6, 6, 2],
    [-6, 9, 2, 6]
])

# 仅求特征值
eigenvals_dict = A.eigenvals()
print("特征值:", list(eigenvals_dict.keys()))

P_cols = []
D_vals = []

# 对每个特征值 lambda，先求 Nul(A - lambda*I) 的基，再标准正交化
for lam in eigenvals_dict.keys():
    mat = A - lam * sp.eye(4)
    null_basis = mat.nullspace()
    ortho_basis = sp.GramSchmidt(null_basis, True)

    for vec in ortho_basis:
        P_cols.append(vec)
        D_vals.append(lam)
        print(f"\nλ = {lam} 的单位特征向量: {vec.T}")

# 构造正交矩阵 P 和对角矩阵 D
P = sp.Matrix.hstack(*P_cols)
D = sp.diag(*D_vals)

print("\n正交矩阵 P =")
sp.pprint(P)
print("\n对角矩阵 D =")
sp.pprint(D)

# 验证 P^T A P = D 且 P^T P = I
print("\n验证 P^T * P 是否为 I:", P.T * P == sp.eye(4))
print("验证 P * D * P^T 是否等于 A:", P * D * P.T == A)
```

## 7.2 Quadratic Forms

**二次型的定义**：$\mathbb R^n$ 上的二次型是形如

$$
Q(\mathbf x)=\mathbf x^TA\mathbf x
$$

的函数，其中 $A$ 为 $n\times n$ 对称矩阵，称为二次型的矩阵。写成分量形式：

$$
Q(\mathbf x)=\sum_{i,j=1}^n a_{ij}x_ix_j.
$$

由于 $A$ 对称，交叉项 $x_ix_j$ 的系数由 $a_{ij}$ 与 $a_{ji}$ 平分。例如

$$
Q(x_1,x_2)=x_1^2-8x_1x_2-5x_2^2
$$

的矩阵为

$$
A=\begin{bmatrix}
1&-4\\
-4&-5
\end{bmatrix}.
$$

### Change of Variable in a Quadratic Form

**定理 4（主轴定理）**：设 $A$ 是 $n\times n$ 实对称矩阵，$Q(\mathbf x)=\mathbf x^TA\mathbf x$。则存在正交矩阵 $P$，使换元 $\mathbf x=P\mathbf y$ 满足

$$
Q(\mathbf x)=\mathbf x^TA\mathbf x
=(P\mathbf y)^TA(P\mathbf y)
=\mathbf y^T(P^TAP)\mathbf y
=\mathbf y^TD\mathbf y
=\lambda_1y_1^2+\cdots+\lambda_ny_n^2,
$$

其中 $D=P^TAP$ 是对角矩阵，$\lambda_1,\dots,\lambda_n$ 是 $A$ 的特征值。

也就是说，任一二次型都能通过正交换元消去交叉项，新矩阵为 $P^TAP$。

### 主轴的含义

正交矩阵 $P$ 的列向量是 $A$ 的单位特征向量，称为二次型

$$
\mathbf x^TA\mathbf x
$$

的**主轴**，$\mathbf y$ 是 $\mathbf x$ 在这组正交主轴下的坐标。


**几何意义**：对 $2\times2$ 对称矩阵，方程

$$
\mathbf x^TA\mathbf x=c
$$

通常表示椭圆、双曲线、两条相交直线、单点或空集。若 $A$ 不是对角矩阵，图形相对于标准坐标轴是“斜的”；主轴就是使图形回到标准位置的新坐标轴。

**二次型的分类**：

- **正定**：$\mathbf x^TA\mathbf x>0$ 对所有 $\mathbf x\neq\mathbf 0$；
- **负定**：$\mathbf x^TA\mathbf x<0$ 对所有 $\mathbf x\neq\mathbf 0$；
- **不定**：既能取正值又能取负值；
- **半正定**：$\mathbf x^TA\mathbf x\ge0$ 对所有 $\mathbf x$；
- **半负定**：$\mathbf x^TA\mathbf x\le0$ 对所有 $\mathbf x$。

正定矩阵指二次型正定的对称矩阵，其他术语类似。

**定理 5**：设 $A$ 对称，则

a. $Q$ 正定 $\iff$ $A$ 的所有特征值 $>0$；  
b. $Q$ 负定 $\iff$ $A$ 的所有特征值 $<0$；  
c. $Q$ 不定 $\iff$ $A$ 有正有负特征值。

证明思路：正交换元后

$$
Q=\lambda_1y_1^2+\cdots+\lambda_ny_n^2,
$$

符号完全由 $\lambda_i$ 决定。

**数值注意：Cholesky 分解**：判断对称矩阵 $A$ 是否正定，可用 Cholesky 分解

$$
A=R^TR,
$$

其中 $R$ 为上三角且对角元为正。$A$ 正定 $\iff$ 这种分解存在。

### 练习 7.2 15

$$
Q=-3x_1^2-7x_2^2-10x_3^2-10x_4^2
+4x_1x_2+4x_1x_3+4x_1x_4+6x_3x_4.
$$

矩阵：

$$
A=
\begin{bmatrix}
-3&2&2&2\\
2&-7&0&0\\
2&0&-10&3\\
2&0&3&-10
\end{bmatrix}.
$$


```python
import sympy as sp

A = sp.Matrix([
    [-3,  2,  2,  2],
    [ 2, -7,  0,  0],
    [ 2,  0,-10,  3],
    [ 2,  0,  3,-10]
])

print("A =")
sp.pprint(A)

lam = sp.symbols('lambda')
p = A.charpoly(lam).as_expr()
sp.pprint(sp.expand(p))
print("因式分解 =")
sp.pprint(sp.factor(p))

eig = A.eigenvals()
print("\n特征值（含重数）:")
sp.pprint(eig)

# 按降序排列特征值
eig_sorted = sorted(eig.keys(), key=lambda x: float(x), reverse=True)
print("\n按降序排列的特征值:", eig_sorted)

all_neg = all(sp.re(ev) < 0 for ev in eig_sorted)
all_pos = all(sp.re(ev) > 0 for ev in eig_sorted)
has_zero = any(sp.simplify(ev) == 0 for ev in eig_sorted)

if all_pos and not has_zero:
    print("\n分类：正定")
elif all_neg and not has_zero:
    print("\n分类：负定")
elif all_pos and has_zero:
    print("\n分类：半正定（不正定）")
elif all_neg and has_zero:
    print("\n分类：半负定（不负定）")
else:
    print("\n分类：不定")

P_cols = []
D_vals = []

for ev in eig_sorted:
    # 构造 A - λI
    M = A - ev * sp.eye(4)
    # 求零空间基
    ns = M.nullspace()
    # 若 eigenspace 维数 > 1，用 Gram-Schmidt 标准正交化
    if len(ns) > 1:
        ortho = sp.GramSchmidt(ns, True)
    else:
        v = ns[0]
        ortho = [v / v.norm()]
    for u in ortho:
        P_cols.append(sp.simplify(u))
        D_vals.append(ev)
        print(f"\nλ = {ev} 对应的单位特征向量:")
        sp.pprint(u.T)

P = sp.Matrix.hstack(*P_cols)
D = sp.diag(*D_vals)

print("\n正交矩阵 P =")
sp.pprint(P)
print("\n对角矩阵 D =")
sp.pprint(D)

```

### 练习 7.2 32

设
$$
A=\begin{bmatrix}a&b\\b&d\end{bmatrix},\qquad \det A\ne 0,
$$
$Q(\mathbf x)=\mathbf x^TA\mathbf x$。验证下列结论：

a. 若 $\det A>0$ 且 $a>0$，则 $Q$ 正定。

b. 若 $\det A>0$ 且 $a<0$，则 $Q$ 负定。

c. 若 $\det A<0$，则 $Q$ 不定。



**解：**

设

$$
A=\begin{bmatrix}a&b\\b&d\end{bmatrix},\qquad \det A\neq0.
$$

特征值为 $\lambda_1,\lambda_2$。由特征多项式：

$$
\lambda_1+\lambda_2=a+d,\qquad
\lambda_1\lambda_2=\det A=ad-b^2.
$$

**(a)** 若 $\det A>0$ 且 $a>0$，则 $ad-b^2>0$，所以 $ad>0$，$a,d$ 同号，且 $d>0$。于是

$$
\lambda_1\lambda_2>0,\qquad
\lambda_1+\lambda_2=a+d>0.
$$

同号且和正，故 $\lambda_1,\lambda_2>0$，$Q$ 正定。

**(b)** 若 $\det A>0$ 且 $a<0$，同理 $d<0$，$\lambda_1+\lambda_2<0$，且乘积正，故两特征值均负，$Q$ 负定。

**(c)** 若 $\det A<0$，则 $\lambda_1\lambda_2<0$，两特征值异号，$Q$ 不定。




### 练习 7.2 33

证明：若 $B$ 是 $m\times n$ 矩阵，则 $B^TB$ 半正定；若 $B$ 是 $n\times n$ 可逆矩阵，则 $B^TB$ 正定。




**证明：**

对任意 $\mathbf x$,

$$
\mathbf x^T(B^TB)\mathbf x
=(B\mathbf x)^T(B\mathbf x)
=\|B\mathbf x\|^2\ge0.
$$

所以 $B^TB$ 半正定。

若 $B$ 可逆，则 $\mathbf x\neq\mathbf 0\Rightarrow B\mathbf x\neq\mathbf 0$，于是

$$
\|B\mathbf x\|^2>0.
$$

故 $B^TB$ 正定。

## 7.3 Constrained Optimization 约束优化

$$
\text{约束优化问题}
\;\Longrightarrow\;
\text{无交叉项的特殊情形}
\;\Longrightarrow\;
\text{一般二次型的最大值/最小值}
\;\Longrightarrow\;
\text{加正交约束后的后续特征值}
\;\Longrightarrow\;
\text{应用例子}.
$$


教材先说明：工程师、经济学家、科学家、数学家经常需要求二次型 $Q(\mathbf x)$ 在某个指定集合上的最大值或最小值。通常可以把问题安排成让 $\mathbf x$ 在**单位向量集合**上变化。

单位约束可以写成等价形式：

$$
\|\mathbf x\|=1
\iff
\|\mathbf x\|^2=1
\iff
\mathbf x^T\mathbf x=1
\iff
x_1^2+x_2^2+\cdots+x_n^2=1.
$$

所以本节研究在**单位球面上的二次型**：

$$
\max_{\|\mathbf x\|=1}\mathbf x^TA\mathbf x
\qquad\text{和}\qquad
\min_{\|\mathbf x\|=1}\mathbf x^TA\mathbf x.
$$

因为

$$
\|\mathbf x\|=1
\iff
\|\mathbf x\|^2=1
\iff
\mathbf x^T\mathbf x=1.
$$

正交对角化 $A=PDP^T$，令 $\mathbf x=P\mathbf y$。正交矩阵保持长度：

$$
\|\mathbf x\|=\|P\mathbf y\|=\|\mathbf y\|.
$$

所以约束变为 $\|\mathbf y\|=1$，而

$$
\mathbf x^TA\mathbf x
=\mathbf y^TD\mathbf y
=\lambda_1y_1^2+\cdots+\lambda_ny_n^2.
$$

若

$$
\lambda_1\ge\lambda_2\ge\cdots\ge\lambda_n,
$$

则

$$
\lambda_n\le \mathbf x^TA\mathbf x\le\lambda_1.
$$

最大值在最大特征值对应的单位特征向量处取得，最小值在最小特征值对应的单位特征向量处取得。



定义

$$
m=\min_{\|\mathbf x\|=1}\mathbf x^TA\mathbf x,
\qquad
M=\max_{\|\mathbf x\|=1}\mathbf x^TA\mathbf x.
$$

 **定理6**  
> 设 $A$ 为对称矩阵，则 $M$ 是 $A$ 的最大特征值 $\lambda_1$，$m$ 是 $A$ 的最小特征值 $\lambda_n$。  
> 当 $\mathbf x$ 是对应于 $M$ 的单位特征向量 $\mathbf u_1$ 时，$\mathbf x^TA\mathbf x=M$；  
> 当 $\mathbf x$ 是对应于 $m$ 的单位特征向量时，$\mathbf x^TA\mathbf x=m$。


证明思路：

1. 正交对角化 $A=PDP^T$，令 $\mathbf x=P\mathbf y$；
2. 正交变换保持长度，所以 $\|\mathbf x\|=1\iff\|\mathbf y\|=1$；
3. 
$$
\mathbf x^TA\mathbf x
=\mathbf y^TD\mathbf y
=\lambda_1y_1^2+\cdots+\lambda_ny_n^2;
$$
4. 设 $\lambda_1\ge\cdots\ge\lambda_n$，则
$$
\lambda_n\le \mathbf y^TD\mathbf y\le\lambda_1;
$$
5. 上界在 $\mathbf y=\mathbf e_1$ 时取到，对应 $\mathbf x=\mathbf u_1$；下界在 $\mathbf y=\mathbf e_n$ 时取到，对应 $\mathbf x=\mathbf u_n$。

**定理7：**

> 设 $A,\lambda_1,\mathbf u_1$ 如定理 6。则
> $$
> \max_{\substack{\mathbf x^T\mathbf x=1\\ \mathbf x^T\mathbf u_1=0}}
> \mathbf x^TA\mathbf x
> $$
> 等于第二特征值 $\lambda_2$，且在 $\mathbf x=\mathbf u_2$ 时取得。

直观理解：把 $\mathbf u_1$ 方向排除后，剩下的单位球面上，二次型的最大值就是第二大特征值。



**定理 8**：继续加约束

$$
\mathbf x^T\mathbf u_1=0,\dots,\mathbf x^T\mathbf u_{k-1}=0,
$$

最大值就是第 $k$ 大特征值 $\lambda_k$，在 $\mathbf u_k$ 处取得。

## 7.4 The Singular Value Decomposition 奇异值分解

**为什么需要 SVD**：对称矩阵可以正交对角化，但长方形矩阵不行。SVD 把任意 $m\times n$ 矩阵分解为

$$
A=U\Sigma V^T,
$$

其中 $U,V$ 正交，$\Sigma$ 为“对角”矩阵。几何上：单位球经 $A$ 映射后变成椭球；椭球的半轴长度就是奇异值，方向由左奇异向量给出。

**$A^TA$ 与奇异值**：对任意 $A$，$A^TA$ 对称且半正定：

$$
\mathbf x^T(A^TA)\mathbf x=\|A\mathbf x\|^2\ge0.
$$

因此 $A^TA$ 的特征值非负。设

$$
\lambda_1\ge\lambda_2\ge\cdots\ge\lambda_n\ge0.
$$

奇异值定义为

$$
\sigma_i=\sqrt{\lambda_i}.
$$

若 $\mathbf v_i$ 是 $A^TA$ 的单位特征向量，则

$$
\|A\mathbf v_i\|^2
=\mathbf v_i^TA^TA\mathbf v_i
=\lambda_i
=\sigma_i^2.
$$

所以奇异值是 $A$ 把单位向量拉长/压缩的倍数。



### **定理 9（$A\mathbf v_i$ 构成列空间正交基）**

设 $\{\mathbf v_1,\dots,\mathbf v_n\}$ 是 $A^TA$ 的标准正交特征向量基，对应特征值降序排列。若 $A$ 有 $r$ 个非零奇异值，则

$$
\{A\mathbf v_1,\dots,A\mathbf v_r\}
$$

是 $\operatorname{Col}A$ 的正交基，且

$$
\operatorname{rank}A=r.
$$

证明：

任取两个不同的向量 $A\mathbf{v}_i$ 和 $A\mathbf{v}_j$（$i \neq j$），算它们的内积：
$$
 (A\mathbf{v}_i)^T(A\mathbf{v}_j) = \mathbf{v}_i^T A^T A \mathbf{v}_j 
$$
因为 $A^TA\mathbf{v}_j = \lambda_j \mathbf{v}_j$，代入得：
$$
 \mathbf{v}_i^T (\lambda_j \mathbf{v}_j) = \lambda_j (\mathbf{v}_i^T \mathbf{v}_j) 
$$
因为 $\mathbf{v}_i$ 和 $\mathbf{v}_j$ 是正交的，内积为 0，所以结果必然是 0。这组向量天然互相垂直。


算一下 $A\mathbf{v}_i$ 的长度平方：
$$
 \|A\mathbf{v}_i\|^2 = \mathbf{v}_i^T A^T A \mathbf{v}_i = \lambda_i 
$$
因为奇异值是 $\sqrt{\lambda_i}$。已知 $A$ 有 $r$ 个非零奇异值，这就意味着前 $r$ 个 $A\mathbf{v}_i$ 的长度不为 0，而后面的全是零向量。去掉零向量，就剩下了 $r$ 个互相垂直的非零向量，它们必然线性无关。

任取列空间里的一个向量 $\mathbf{y}$，它肯定能写成 $\mathbf{y} = A\mathbf{x}$。
把 $\mathbf{x}$ 按基 $\mathbf{v}_1, \dots, \mathbf{v}_n$ 展开：
$$
 \mathbf{y} = A(c_1\mathbf{v}_1 + \dots + c_n\mathbf{v}_n) 
$$
乘进去：
$$
 \mathbf{y} = c_1A\mathbf{v}_1 + \dots + c_rA\mathbf{v}_r + c_{r+1}A\mathbf{v}_{r+1} + \dots + c_nA\mathbf{v}_n 
$$
因为后面 $n-r$ 个 $A\mathbf{v}_i$ 全是零向量，所以：
$$
 \mathbf{y} = c_1A\mathbf{v}_1 + \dots + c_rA\mathbf{v}_r 
$$
这说明，列空间里的**任何**向量，都能被前 $r$ 个向量线性表出。

**结论**：这 $r$ 个向量既两两垂直、又非零、还能凑出整个列空间。所以 $\{A\mathbf{v}_1, \dots, A\mathbf{v}_r\}$ 就是列空间的正交基。基里有 $r$ 个向量，维数就是 $r$，所以 $\operatorname{rank}A = r$。


**数值注意**：矩阵的秩对元素的小改动可能很敏感。用行化简数主元列并不可靠：舍入误差会让一个本来不满秩的矩阵在化简后看起来满秩。

实际计算中，估计大型矩阵的秩最可靠的办法是数非零奇异值的个数，极小的奇异值按 0 处理。去掉这些极小值后剩下的非零奇异值个数，称为矩阵的**有效秩**。



### **定理10：SVD 定理**

设 $A$ 为 $m\times n$，秩为 $r$。则存在正交矩阵 $U$（$m\times m$）、$V$（$n\times n$）和 $m\times n$ 的 $\Sigma$，使得

$$
A=U\Sigma V^T,
$$

其中

$$
\Sigma=
\begin{bmatrix}
D&0\\
0&0
\end{bmatrix},
\qquad
D=\operatorname{diag}(\sigma_1,\dots,\sigma_r),
$$

且

$$
\sigma_1\ge\cdots\ge\sigma_r>0.
$$

$U$ 的列称为左奇异向量，$V$ 的列称为右奇异向量。


证明：

设 $A$ 是 $m\times n$ 矩阵，秩为 $r$。

取 $A^TA$ 的标准正交特征向量基 $\{\mathbf v_1,\dots,\mathbf v_n\}$，对应特征值 $\lambda_1\ge\cdots\ge\lambda_r>0$，$\lambda_{r+1}=\cdots=\lambda_n=0$，并令 $\sigma_i=\sqrt{\lambda_i}$。由定理 9，$\{A\mathbf v_1,\dots,A\mathbf v_r\}$ 是 $\operatorname{Col}A$ 的正交基，且 $\|A\mathbf v_i\|=\sigma_i$。

把它们单位化。令

$$
\mathbf u_i=\frac{1}{\sigma_i}A\mathbf v_i\qquad(1\le i\le r),
$$

即 $A\mathbf v_i=\sigma_i\mathbf u_i$，此时 $\{\mathbf u_1,\dots,\mathbf u_r\}$ 是 $\operatorname{Col}A$ 的标准正交基。

补足 $U$。$U$ 必须是 $m\times m$ 矩阵，而现在只有 $r$ 个向量，$r\le m$。把这组标准正交向量扩充为 $\mathbb R^m$ 的标准正交基 $\{\mathbf u_1,\dots,\mathbf u_m\}$，令 $U=[\mathbf u_1\ \cdots\ \mathbf u_m]$。再令 $V=[\mathbf v_1\ \cdots\ \mathbf v_n]$，它是 $n\times n$ 正交矩阵；$\Sigma$ 与 $A$ 同尺寸，左上角放奇异值，其余补零：

$$
\Sigma=\begin{bmatrix}D&0\\0&0\end{bmatrix},
\qquad
D=\operatorname{diag}(\sigma_1,\dots,\sigma_r).
$$

比较 $AV$ 与 $U\Sigma$ 的列。当 $i\le r$ 时 $A\mathbf v_i=\sigma_i\mathbf u_i$；当 $i>r$ 时 $\sigma_i=0$，故 $A\mathbf v_i=\mathbf 0$。于是

$$
AV=[\sigma_1\mathbf u_1\ \cdots\ \sigma_r\mathbf u_r\ \mathbf 0\ \cdots\ \mathbf 0]=U\Sigma.
$$

又 $V$ 正交，$VV^T=I$，所以

$$
U\Sigma V^T=AVV^T=A.
$$


**三步构造法**：

第 1 步，正交对角化 $A^TA$，求它的特征值和特征向量。

第 2 步，构造 $V$ 和 $\Sigma$。$V$ 的列是 $A^TA$ 的单位特征向量，按特征值降序排列；$\Sigma$ 的对角元是奇异值 $\sigma_i=\sqrt{\lambda_i}$，其余位置补零。

第 3 步，构造 $U$。前 $r$ 列取 $\mathbf u_i=\frac{1}{\sigma_i}A\mathbf v_i$。若 $r<m$，还需要补 $m-r$ 列：设待求向量为 $\mathbf x=(x_1,\dots,x_m)^T$，要求 $\mathbf u_i^T\mathbf x=0$（$i=1,\dots,r$），解这个齐次方程组得到基础解系，再用 Gram-Schmidt 正交化、单位化，作为 $U$ 的后 $m-r$ 列。


### Applications of the Singular Value Decomposition 奇异值分解的应用
**四个基本子空间**：若

$$
A=U\Sigma V^T,\qquad \operatorname{rank}A=r,
$$

则：

- 列空间$\operatorname{Col}A$ 的标准正交基：$\mathbf u_1,\dots,\mathbf u_r$；
- 左零空间$\operatorname{Nul}A^T$ 的标准正交基：$\mathbf u_{r+1},\dots,\mathbf u_m$；
- 行空间$\operatorname{Row}A$ 的标准正交基：$\mathbf v_1,\dots,\mathbf v_r$；
- 零空间$\operatorname{Nul}A$ 的标准正交基：$\mathbf v_{r+1},\dots,\mathbf v_n$。
![7-1.png](picture/chapter_7_picture/7-1.png)

这可视为可逆矩阵定理的补充：

$$
A\ \text{可逆}
\iff
A\ \text{有}\ n\ \text{个非零奇异值}.
$$


**可逆矩阵定理（结论部分）**

设 $A$ 是一个 $n \times n$ 矩阵,下列说法都与“$A$ 是可逆矩阵”这一命题等价：

**s.** $(\operatorname{Col} A)^{\perp} = \{\mathbf{0}\}$。

**t.** $(\operatorname{Nul} A)^{\perp} = \mathbb{R}^n$。

**u.** $\operatorname{Row} A = \mathbb{R}^n$。

**v.** $A$ 有 $n$ 个非零奇异值。





**约化 SVD 与伪逆**：令

$$
U_r=[\mathbf u_1\ \cdots\ \mathbf u_r],\qquad
V_r=[\mathbf v_1\ \cdots\ \mathbf v_r].
$$

则

$$
A=U_rDV_r^T.
$$

伪逆定义为

$$
A^+=V_rD^{-1}U_r^T.
$$

伪逆给出最小长度最小二乘解：

$$
\hat{\mathbf x}=A^+\mathbf b.
$$

**条件数与数值注意**：若 $A$ 可逆，条件数为

$$
\kappa_2(A)=\frac{\sigma_1}{\sigma_n}.
$$

条件数越大，$A\mathbf x=\mathbf b$ 对误差越敏感。实际计算 SVD 时通常避免显式构造 $A^TA$，因为会平方误差；现代算法用迭代方法直接算奇异值。

### 练习 7.4 27

**27.** 计算下列矩阵的一个 SVD，最终矩阵元素保留两位小数，使用例 3 和例 4 的方法。
$$
A=
\begin{bmatrix}
6&-8&-4&5&-4\\
2&7&-5&-6&4\\
0&-1&-8&2&2\\
-1&-2&4&4&-8
\end{bmatrix}.
$$



```python
import numpy as np

def custom_svd(A):
    m, n = A.shape
    
    # 求 A^T A 的特征值和特征向量
    ATA = A.T @ A
    eigenvalues, eigenvectors = np.linalg.eigh(ATA) 
    
    # 降序排列
    idx = eigenvalues.argsort()[::-1]
    eigenvalues = eigenvalues[idx]
    V = eigenvectors[:, idx]
    
    # 过滤掉极小值
    tolerance = 1e-8
    singular_values = np.sqrt(np.maximum(eigenvalues, 0))
    r = min(np.sum(singular_values > tolerance), m, n)
    
    # 构造 V 和 Sigma
    V_reduced = V[:, :r] 
    sigmas = singular_values[:r]
    
    Sigma = np.zeros((m, n))
    for i in range(r):
        Sigma[i, i] = sigmas[i]
    
    # 构造 U 的前 r 列
    U = np.zeros((m, r))
    for i in range(r):
        U[:, i] = (A @ V_reduced[:, i]) / sigmas[i]
        
    # 若 r < m，正交扩充 U
    if r < m:
        for i in range(m):
            if U.shape[1] == m: break
            vec = np.zeros(m)
            vec[i] = 1.0
            for j in range(U.shape[1]):
                vec -= (U[:, j] @ vec) * U[:, j]
            if np.linalg.norm(vec) > tolerance:
                vec = vec / np.linalg.norm(vec)
                U = np.hstack((U, vec.reshape(-1, 1)))
                
    return U, Sigma, V.T

# 测试矩阵
A = np.array([
    [6, -8, -4, 5, -4],
    [2, 7, -5, -6, 4],
    [0, -1, -8, 2, 2],
    [-1, -2, 4, 4, -8]
])

U, Sigma, Vt = custom_svd(A)

# 打印结果
np.set_printoptions(precision=2, suppress=True, floatmode='fixed')
print("左奇异矩阵 U (4x4):\n", U)
print("\n对角矩阵 Sigma (4x5):\n", Sigma)
print("\n右奇异矩阵的转置 V^T (5x5):\n", Vt)

# 验证 A = U * Sigma * V^T
A_reconstructed = U @ Sigma @ Vt
print("\n验证还原的矩阵 A:\n", A_reconstructed)
```

## 7.5 Applications to Image Processing and Statistics 应用于图像处理与统计

这一节把多维观测数据看成空间中的一堆点：先中心化，用协方差矩阵描述这堆点的形状，再通过正交对角化换坐标，使新变量彼此不相关、方差依次最大，最后只保留前几个方差最大的方向，完成降维和信息压缩。图像处理是其中一类典型应用。

一个观测对象不再是一个数，而是一个向量。若每个样本测 $p$ 个指标，就得到 $N$ 个观测向量

$$
\mathbf X_1,\dots,\mathbf X_N\in\mathbb R^p,
$$

把它们按列排成 $p\times N$ 矩阵

$$
[\mathbf X_1\ \cdots\ \mathbf X_N].
$$

也就是说，矩阵的列是样本，行是变量。



### **样本均值与中心化**

设
$$
\mathbf M=\frac1N(\mathbf X_1+\cdots+\mathbf X_N)
$$
是样本均值向量。它相当于散点图的“中心”。

然后令
$$
\hat{\mathbf X}_k=\mathbf X_k-\mathbf M.
$$
把每个观测向量减去均值，得到
$$
B=[\hat{\mathbf X}_1\ \cdots\ \hat{\mathbf X}_N].
$$
这时 $B$ 的列均值为零，称为**均值偏差形式**（mean-deviation form）。

为什么要中心化？因为 PCA 关心的是数据“围绕均值怎么变化”，而不是数据中心在哪里。


### **协方差矩阵**

定义样本协方差矩阵

$$
S=\frac1{N-1}BB^T.
$$

由于

$$
BB^T=\sum_{k=1}^N \hat{\mathbf X}_k\hat{\mathbf X}_k^T,
$$

所以 $S$ 是 $p\times p$ 对称矩阵，并且半正定。

对角元 $s_{jj}$ 是第 $j$ 个变量的方差，衡量这个变量的离散程度；非对角元 $s_{ij}$ 是第 $i$ 个变量与第 $j$ 个变量的协方差，$s_{ij}=0$ 时称两个变量不相关。对角线元素之和 $\operatorname{tr}(S)$ 是数据的总方差。




### **PCA**

现在要找正交矩阵
$$
P=[\mathbf u_1\ \cdots\ \mathbf u_p],
$$
做变量替换
$$
\mathbf X=P\mathbf Y,
$$
等价地
$$
\mathbf Y=P^T\mathbf X.
$$
这里 $\mathbf X$ 是原始变量向量，$\mathbf Y$ 是新变量向量。

对每个观测 $\mathbf X_k$，新坐标是
$$
\mathbf Y_k=P^T\mathbf X_k.
$$
教材指出：若原数据协方差为 $S$，则新数据 $\mathbf Y_1,\dots,\mathbf Y_N$ 的协方差矩阵是
$$
P^TSP.
$$
所以 PCA 的目标就是选正交 $P$，使得
$$
P^TSP
$$
是对角矩阵。这样新变量 $y_1,\dots,y_p$ 彼此不相关。

因为 $S$ 对称，由 7.1 节可正交对角化：
$$
S=PDP^T
$$
其中 $P$ 的列是 $S$ 的单位特征向量，$D$ 的对角元是特征值。把特征值按
$$
\lambda_1\ge \lambda_2\ge\cdots\ge\lambda_p\ge 0
$$
排列，对应单位特征向量为
$$
\mathbf u_1,\dots,\mathbf u_p.
$$
这些 $\mathbf u_j$ 就叫数据的**主成分**：

- 第一主成分 $\mathbf u_1$：对应最大特征值 $\lambda_1$；
- 第二主成分 $\mathbf u_2$：对应第二大特征值 $\lambda_2$；
- 以此类推。

新变量为
$$
y_j=\mathbf u_j^T\mathbf X.
$$
若 $\mathbf u_1=(c_1,\dots,c_p)^T$，则
$$
y_1=c_1x_1+\cdots+c_px_p,
$$
即第一主成分是原始变量的一个线性组合，权重就是第一特征向量的分量。



### **方差解释与降维**

因为
$$
P^TSP=D,
$$
所以新变量 $y_j$ 的方差就是 $\lambda_j$。总方差不变：
$$
\text{总方差}=\operatorname{tr}(S)=\lambda_1+\cdots+\lambda_p.
$$
第 $j$ 个主成分解释的方差比例是
$$
\frac{\lambda_j}{\lambda_1+\cdots+\lambda_p}.
$$

如果前几个特征值占了总方差的绝大部分，就可以只保留前几个主成分，实现降维。



**为什么第一主成分方差最大：与 7.3 的联系**

对任意单位向量 $\mathbf u$，变量

$$
y=\mathbf u^T\mathbf X
$$

的方差是

$$
\mathbf u^TS\mathbf u.
$$

由 7.3 节定理 8，$\mathbf u^TS\mathbf u$ 在 $\|\mathbf u\|=1$ 下的最大值是 $S$ 的最大特征值 $\lambda_1$，在 $\mathbf u=\mathbf u_1$ 时达到。第二主成分是在与第一主成分不相关的约束下使方差最大的方向，对应 $\lambda_2$，依此类推。

所以 PCA 是按方差最大、且彼此不相关的顺序依次选方向，而不是随便取线性组合。



**数值实现：与 7.4 SVD 的联系**

7.5 末尾 Numerical Notes 说明实际计算常用 SVD，而不是直接对 $S$ 做特征分解。

若 $B$ 是中心化后的 $p\times N$ 矩阵，令
$$
A=\frac1{\sqrt{N-1}}B^T.
$$
则
$$
A^TA=\frac1{N-1}BB^T=S.
$$
于是：

- $A$ 的奇异值的平方就是 $S$ 的特征值；
- $A$ 的右奇异向量就是主成分；
- 用 SVD 计算比直接构造 $S$ 再特征分解更稳定、更快.

这正好把 7.4 的 SVD 和 7.5 的 PCA 连起来。



这一节用到前面三块工具：

- 7.1：对称矩阵的正交对角化，用来把 $S$ 化成对角矩阵；
- 7.3：约束下二次型的最大值，用来说明第一主成分方差最大；
- 7.4：SVD，作为协方差矩阵特征分解的稳定数值实现。

> $$

### Supplementary Exercise 28

**28.** 设 $A=U_rDV_r^T$ 是 $m\times n$ 矩阵 $A$ 的约化 SVD，伪逆为
$$
A^+=V_rD^{-1}U_r^T.
$$
验证 $A^+$ 的下列性质：

a. 对每个 $\mathbf y\in\mathbb R^m$，$AA^+\mathbf y$ 是 $\mathbf y$ 在 $\operatorname{Col}A$ 上的正交投影。

b. 对每个 $\mathbf x\in\mathbb R^n$，$A^+A\mathbf x$ 是 $\mathbf x$ 在 $\operatorname{Row}A$ 上的正交投影。

c. $AA^+A=A$，且 $A^+AA^+=A^+$。


**解：**

已知约化 SVD 为 $A=U_rDV_r^T$，伪逆为 $A^+=V_rD^{-1}U_r^T$。

$U_r$ 和 $V_r$ 的列向量是标准正交的，即 $U_r^TU_r=I_r$，$V_r^TV_r=I_r$。

**a.**
计算 $AA^+$：
$$
AA^+ = (U_rDV_r^T)(V_rD^{-1}U_r^T) = U_rD(V_r^TV_r)D^{-1}U_r^T = U_rI_rD^{-1}U_r^T = U_rU_r^T
$$
由于 $U_r^TU_r=I_r$，矩阵 $U_rU_r^T$ 满足：
1. 对称性：$(U_rU_r^T)^T = U_rU_r^T$
2. 幂等性：$(U_rU_r^T)^2 = U_r(U_r^TU_r)U_r^T = U_rU_r^T$

因此 $U_rU_r^T$ 是正交投影矩阵。

又因为 $A$ 的列空间 $\operatorname{Col}A = \operatorname{Col}U_r$，所以 $AA^+\mathbf y = U_rU_r^T\mathbf y$ 是 $\mathbf y$ 在 $\operatorname{Col}A$ 上的正交投影。



**b.**
计算 $A^+A$：
$$
A^+A = (V_rD^{-1}U_r^T)(U_rDV_r^T) = V_rD^{-1}(U_r^TU_r)DV_r^T = V_rD^{-1}I_rDV_r^T = V_rV_r^T
$$
同理，$V_rV_r^T$ 满足对称性和幂等性，是一个正交投影矩阵。由于 $A$ 的行空间 $\operatorname{Row}A = \operatorname{Col}(A^T) = \operatorname{Col}V_r$，所以 $A^+A\mathbf x = V_rV_r^T\mathbf x$ 是 $\mathbf x$ 在 $\operatorname{Row}A$ 上的正交投影。



**c.**
利用前面算出的 $AA^+=U_rU_r^T$ 和 $A^+A=V_rV_r^T$：
$$
AA^+A = (AA^+)A = (U_rU_r^T)(U_rDV_r^T) = U_r(U_r^TU_r)DV_r^T = U_rI_rDV_r^T = U_rDV_r^T = A
$$
$$
A^+AA^+ = (A^+A)A^+ = (V_rV_r^T)(V_rD^{-1}U_r^T) = V_r(V_r^TV_r)D^{-1}U_r^T = V_rI_rD^{-1}U_r^T = V_rD^{-1}U_r^T = A^+
$$
得证。

### Supplementary Exercise 31

**31.** 构造 $A$ 的伪逆。先用矩阵程序求 $A$ 的 SVD，或先对 $A^TA$ 作正交对角化。用伪逆求解 $A\mathbf x=\mathbf b$，其中
$$
\mathbf b=(6,-1,-4,6),
$$
设解为 $\hat{\mathbf x}$。计算验证 $\hat{\mathbf x}\in\operatorname{Row}A$。求一个非零向量 $\mathbf u\in\operatorname{Nul}A$，并验证
$$
\|\hat{\mathbf x}\|<\|\hat{\mathbf x}+\mathbf u\|,
$$
这由第 29(c) 题必然成立。
$$
A=
\begin{bmatrix}
-3&-3&-6&6&1\\
-1&-1&-1&1&-2\\
0&0&-1&1&-1\\
0&0&-1&1&-1
\end{bmatrix}.
$$



```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)

A = np.array([
    [-3, -3, -6, 6, 1],
    [-1, -1, -1, 1, -2],
    [0,  0, -1, 1, -1],
    [0,  0, -1, 1, -1]
], dtype=float)

b = np.array([6, -1, -4, 6], dtype=float)

U, s, Vt = np.linalg.svd(A, full_matrices=False)


tol = 1e-10
r = np.sum(s > tol)

# 构造伪逆 A^+
U_r = U[:, :r]           # m x r
D_inv = np.diag(1 / s[:r]) # r x r
V_r = Vt[:r, :].T        # n x r
A_plus = V_r @ D_inv @ U_r.T

#  用伪逆求解 Ax = b
x_hat = A_plus @ b

# 验证 x_hat ∈ Row(A)
proj_x_hat = V_r @ (V_r.T @ x_hat)
is_in_row_space = np.allclose(x_hat, proj_x_hat)

#  求一个非零向量 u ∈ Nul(A)
# Nul(A) 由 V 中对应奇异值为 0 的列向量张成 (即 Vt 的后 n-r 行)
V_null = Vt[r:, :].T 
u = V_null[:, 0] # 取零空间基的第一个向量

#  验证 ||x_hat|| < ||x_hat + u||
norm_x_hat = np.linalg.norm(x_hat)
norm_x_hat_plus_u = np.linalg.norm(x_hat + u)


print("矩阵的秩 r =", r)
print("\n--- 求解结果 ---")
print("伪逆 A^+ =\n", A_plus)
print("\n解 x_hat =\n", x_hat)
print("\n--- 验证 ---")
print(f"x_hat 是否在 Row(A) 中? : {is_in_row_space}")
print("\n零空间向量 u =\n", u)
print(f"\n||x_hat|| = {norm_x_hat:.4f}")
print(f"||x_hat + u|| = {norm_x_hat_plus_u:.4f}")
print(f"是否满足 ||x_hat|| < ||x_hat + u|| ? : {norm_x_hat < norm_x_hat_plus_u}")
```
