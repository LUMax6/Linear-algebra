## 5.1 Eigenvectors and Eigenvalues（特征向量与特征值）

### 定义

> **定义**：$A$ 是 $n\times n$ 矩阵，**特征向量（eigenvector）** 是非零向量 $x$，使得存在标量 $\lambda$ 满足
> $$
> Ax=\lambda x
> $$
> $\lambda$ 称为 **特征值（eigenvalue）**；$x$ 称为对应于 $\lambda$ 的特征向量。

- 特征向量**必须非零**（由定义），但特征值**可以为零**。
- $Ax=\lambda x \;\Longleftrightarrow\; (A-\lambda I)x=0$。因此：
  > $\lambda$ 是 $A$ 的特征值 $\iff$ 方程 $(A-\lambda I)x=0$ 有非平凡解 $\iff$ $A-\lambda I$ 不可逆。
- 方程 $(A-\lambda I)x=0$ 的解集就是 $A-\lambda I$ 的零空间，称为**特征空间（eigenspace）**，它是 $\mathbb{R}^n$ 的子空间（含零向量和所有对应特征向量）。
- 行化简**不能用来求特征值**（$A$ 的阶梯形通常不显示特征值）；但已知特征值 $\lambda$ 后，可用行化简求对应特征空间的一组基。

**判断一个给定向量是否特征向量**：直接算 $Ax$，看是否为 $x$ 的倍数。
**判断一个标量是否特征值**：行化简 $(A-\lambda I\mid 0)$，看是否有自由变量（或看 $A-\lambda I$ 是否不可逆）。



### 定理 1（三角矩阵的特征值）

> 三角矩阵的特征值等于其主对角线上的元素。

证明要点（上三角）：$A-\lambda I$ 仍为上三角，其主对角线为 $a_{11}-\lambda,\dots,a_{nn}-\lambda$。$(A-\lambda I)x=0$ 有自由变量 $\iff$ 对角线上至少一个元素为零 $\iff$ $\lambda$ 等于某个 $a_{ii}$。

### 定理 2（不同特征值对应特征向量线性无关）

> 若 $v_1,\dots,v_r$ 是对应于 $n\times n$ 矩阵 $A$ 的**互不相同**特征值 $\lambda_1,\dots,\lambda_r$ 的特征向量，则 $\{v_1,\dots,v_r\}$ 线性无关。

证明思路（反证）：

设线性相关，取最小下标 $p$ 使 $v_{p+1}=c_1v_1+\cdots+c_pv_p$。

两边左乘 $A$ 得 $c_1\lambda_1v_1+\cdots+c_p\lambda_pv_p=\lambda_{p+1}v_{p+1}$

再把前式两边乘 $\lambda_{p+1}$ 后相减，得 $\sum c_i(\lambda_i-\lambda_{p+1})v_i=0$。

由 $\{v_1,\dots,v_p\}$ 线性无关及特征值互异，所有 $c_i=0$，矛盾。

### 特征值与差分方程

一阶差分方程 $x_{k+1}=Ax_k$（$k=0,1,2,\dots$）。若取特征向量 $x_0$ 及对应特征值 $\lambda$，则
$$
x_k=\lambda^k x_0
$$
是方程的解，因为 $Ax_k=A(\lambda^kx_0)=\lambda^k(Ax_0)=\lambda^{k+1}x_0=x_{k+1}$。线性组合仍是解（见作业 5.1 #41）。

- 但根据空间性质， $n $ 维空间里最多只能有  $n $ 个线性无关的向量。

**练习5.1 33**：设 $\lambda$ 是可逆矩阵 $A$ 的特征值，证明 $\lambda^{-1}$ 是 $A^{-1}$ 的特征值。

**练习5.1 35**：证明 $\lambda$ 是 $A$ 的特征值 $\iff$ $\lambda$ 是 $A^T$ 的特征值。

**证明：**
 $\lambda $ 是  $A $ 的特征值  $\iff $ 存在非零向量  $\mathbf{v} $ 使得  $(A - \lambda I)\mathbf{v} = \mathbf{0} $ 有非平凡解  $\iff $ 矩阵  $A - \lambda I $ **不可逆**。

矩阵转置不改变其秩： $\text{rank}(A - \lambda I) = \text{rank}((A - \lambda I)^T) $。

矩阵可逆（满秩）当且仅当其转置可逆。所以： $A - \lambda I $ 不可逆  $\iff $  $(A - \lambda I)^T $ 不可逆。

$(A - \lambda I)^T = A^T - \lambda I $。

$A - \lambda I $ 不可逆  $\iff $  $A^T - \lambda I $ 不可逆。

$A^T - \lambda I $ 不可逆  $\iff $ 存在非零向量  $\mathbf{u} $ 使得  $(A^T - \lambda I)\mathbf{u} = \mathbf{0} $ 有非平凡解  $\iff $  $\lambda $ 是  $A^T $ 的特征值。

**结论：**
 $\lambda $ 是  $A $ 的特征值  $\iff $  $\lambda $ 是  $A^T $ 的特征值。


## 5.2 The Characteristic Equation（特征方程）

>核心目标：把“求矩阵特征值”的问题，转化为“求一个多项式方程的根”的问题。

### 特征多项式与特征方程

> 标量 $\lambda$ 是 $n\times n$ 矩阵 $A$ 的特征值 $\iff$ $\lambda$ 满足**特征方程**
> $$
> \det(A-\lambda I)=0
> $$

$\det(A-\lambda I)$ 是 $\lambda$ 的 $n$ 次多项式，称为 $A$ 的**特征多项式（characteristic polynomial）**。它恰好有 $n$ 个根（计重数，允许复数）。

一个特征值 $\lambda$ 的（代数）重数 = 它在特征方程中作为因子的出现次数（作为根的重数）。

求特征值的标准流程：
$$
\det\begin{bmatrix}a-\lambda&b\\c&d-\lambda\end{bmatrix}=(a-\lambda)(d-\lambda)-bc
$$


### 行列式性质回顾（定理 3）

$A,B$ 为 $n\times n$矩阵。
> a. $A$ 可逆 $\iff \det A\neq0$；
> b. $\det AB=(\det A)(\det B)$；
> c. $\det A^T=\det A$； 
> d. $A$ 三角 $\Rightarrow \det A=$ 主对角线元素之积；
> e. 行替换不变号，行交换变号，行倍乘按倍数缩放行列式。

### 可逆矩阵定理（续）

> $n\times n$ 矩阵 $A$ 可逆 $\iff$ 0 不是 $A$ 的特征值。

（因为 0 是特征值 $\iff$ 存在非零 $x$ 使 $Ax=0$ $\iff$ $A$ 奇异。）

### 相似（Similarity）

> **定义**：$A$ 与 $B$ **相似（similar）**，若存在可逆 $P$ 使 $P^{-1}AP=B$（等价于 $A=PBP^{-1}$）。$A\mapsto P^{-1}AP$ 称为**相似变换**。

> 定理4：若 $A$ 与 $B$ 相似，则它们有相同特征多项式，因而有相同特征值（含重数）。

证明：若 $B=P^{-1}AP$，则 $B-\lambda I=P^{-1}(A-\lambda I)P$，两边取行列式并利用 $\det(P^{-1})\det(P)=1$，得 $\det(B-\lambda I)=\det(A-\lambda I)$。

同特征值**不**推出相似（例如 $\begin{bmatrix}2&1\\0&2\end{bmatrix}$ 与 $2I$ 同特征值但不同）；相似**不是**行等价。

**练习5.2 33**：构造随机整数 $4\times4$ 矩阵 $A$，验证 $A$ 与 $A^T$ 有相同特征多项式（相同特征值、相同重数）。它们有相同的特征向量吗？再对 $5\times5$ 重复。


```python
import numpy as np
from numpy import linalg as la

for n in (4, 5):
    A = np.random.randint(-5, 6, (n, n)).astype(float)
    print(f"\n{n}x{n} 矩阵 A =\n{A}")
    wA, vA = la.eig(A)
    wAt, vAt = la.eig(A.T)
    # 特征值是否相同
    sa, sB = np.sort(wA), np.sort(wAt)
    print("特征值一致:", np.allclose(sa, sB))
    # 特征向量一般不同（A 与 A^T 特征向量仅在特殊矩阵下相同）
    print("A 的特征向量:\n", vA)
    print("A^T 的特征向量:\n", vAt)
```


**结论**：$A$ 与 $A^T$ **一定同特征值**，但**特征向量一般不同**（仅当矩阵对称等特殊情形才可能相同）。



  

**练习5.2 35**设 $A=\begin{bmatrix}-6&28&21\\4&-15&-12\\-8&a&25\end{bmatrix}$。对 $a\in\{32,31.9,31.8,32.1,32.2\}$，计算特征多项式与特征值，画出 $p(t)=\det(A-tI)$ 在 $0\le t\le3$ 的图像（尽量同坐标系），说明 $a$ 变化时特征值如何变化。


```python
import numpy as np
from numpy import linalg as la
import matplotlib.pyplot as plt

t = np.linspace(0, 3, 400)
for a in (32, 31.9, 31.8, 32.1, 32.2):
    A = np.array([[-6, 28, 21], [4, -15, -12], [-8, a, 25]], float)
    # 特征多项式 p(t)=det(A - tI)
    p = np.array([la.det(A - t0*np.eye(3)) for t0 in t])
    plt.plot(t, p, label=f"a={a}")
    w = la.eigvals(A)
    print(f"a={a}: 特征值 = {np.round(w,4)}")
plt.axhline(0, color='gray', lw=.5)
plt.legend(); plt.xlabel("t"); plt.ylabel("p(t)=det(A-tI)"); plt.grid()
plt.show()
```


**现象**：$p(t)=0$ 的根（图像与横轴交点）就是特征值。本例在 $0\le t\le3$ 内有一个固定特征值为1；当 $a$ 从 32 略变（31.8、31.9、32.1、32.2），该根不动，但会出现不同的根。这直观表明：**矩阵特征值对元素 $a$ 的微小扰动相当敏感**，即使每个元素只改变 $0.1\sim0.2$，特征值也会明显变化。




## 5.3 Diagonalization（对角化）

### 定义与定理

> **定义**：方阵 $A$ **可对角化（diagonalizable）**，若 $A$ 相似于对角矩阵，即存在可逆 $P$ 与对角阵 $D$ 使 $A=PDP^{-1}$。

> **对角化定理（定理5）**：$n\times n$ 矩阵 $A$ 可对角化 $\iff$ $A$ 有 $n$ 个线性无关的特征向量。

> 事实上 $A=PDP^{-1}$（$D$ 对角）$\iff$ $P$ 的列是 $A$ 的 $n$ 个线性无关特征向量，此时 $D$ 的对角元是对应这些特征向量的特征值。
>

> 换句话说：$A$ 可对角化 $\iff$ $\mathbb{R}^n$ 中存在由特征向量构成的基（**特征向量基**）。

对角化步骤（Example 3）：①求特征值；②对每个特征值求特征空间的一组基，凑齐 $n$ 个线性无关特征向量（关键步骤，凑不齐则不可对角化）；③用这些向量作 $P$ 的列；④按 $P$ 的列顺序把对应特征值放上 $D$ 的对角线。**$P$ 的列与 $D$ 的对角元顺序必须一一对应**。验证：算 $AP$ 与 $PD$ 是否相等（$AP=PD \iff A=PDP^{-1}$，当 $P$ 可逆时）。

> **定理6**：有 $n$ 个不同特征值的 $n\times n$ 矩阵可对角化。
> （非对角化的必要条件）

> **定理7**：设 $A$ 的不同特征值为 $\lambda_1,\dots,\lambda_p$。

>a.（几何重数与代数重数的关系）： 特征空间维数 $\le$ 该特征值的（代数）重数；

特征空间的维数：解方程 $(A-\lambda I)x = 0 $得到的解空间的基础解系的数量（即几何重数）。

特征值的重数：特征多项式 $|A-\lambda I| = 0$ 中，因子 $(\lambda - \lambda_k)$ 的幂次（即代数重数）

**几何重数永远不会超过代数重数**

> b. $A$ 可对角化 $\iff$ 各特征空间维数之和 $=n$ $\iff$ 特征多项式能完全分解为线性因子 **且** 每个特征空间的维数等于其重数；

线性因子：次数为1的多项式因子

代数基本定理：在复数域中，任何 n 次多项式都可以完全分解为 n 个线性因子的乘积。所以在复数域中，该条件总是成立，需要考虑的就是几何重数和代数重数是否相等的问题。

> c. 若可对角化，各特征空间基的并构成 $\mathbb{R}^n$ 的特征向量基。

**对角化用处**：$A=PDP^{-1}\Rightarrow A^k=PD^kP^{-1}$，其中 $D^k=\mathrm{diag}(\lambda_1^k,\dots,\lambda_n^k)$，从而高效计算高次幂。

**练习5.3 39-41**：用矩阵程序的特征值命令求特征值，再按 5.1 的方法求特征空间基并完成对角化。


```python
import sympy as sp

def method_5_1_diagonalize(A, exercise_name):
    print(f"\n{'-'*5} Exercise {exercise_name} {'-'*5}")
    print("原矩阵 A:")
    sp.pprint(A)
    
    # 求特征值
    eigenvals_dict = A.eigenvals()
    print(f"特征值及其代数重数: {eigenvals_dict}")
    
    P_columns = []
    D_diagonal = []
    
    # 对每个特征值求特征空间的基
    for val, mult in eigenvals_dict.items():
        print(f"\n特征值 lambda = {val} (代数重数: {mult})")
        #  构造矩阵 (A - lambda*I)
        A_minus_lambdaI = A - val * sp.eye(A.rows)
        print("矩阵 (A - lambda*I):")
        sp.pprint(A_minus_lambdaI)
        
        # 求矩阵的零空间
        null_space_basis = A_minus_lambdaI.nullspace()
        
        if len(null_space_basis) < mult:
            print(f"几何重数 ({len(null_space_basis)}) < 代数重数 ({mult})，矩阵不可对角化")
        else:
            print(f"求得特征空间的基 (几何重数 = {len(null_space_basis)}):")
            for vec in null_space_basis:
                sp.pprint(vec)
                P_columns.append(vec)
        
        # 记录特征值，放入对角矩阵D
        for _ in range(mult):
             D_diagonal.append(val)
             
    # 并验证对角化
    if len(P_columns) == A.rows:
        P = sp.Matrix.hstack(*P_columns)
        D = sp.diag(*D_diagonal)
        print("矩阵 P:")
        sp.pprint(P)
        print("\n对角矩阵 D :")
        sp.pprint(D)
        
        print("\n验证 A = P * D * P^-1")
        result = P * D * P.inv()
        sp.pprint(result)
    else:
        print("\n特征向量数量不足，该矩阵无法对角化。")

# 题目 39 
A39 = [
    [-6, 4, 0, 9],
    [-3, 0, 1, 6],
    [-1, -2, 1, 0],
    [-4, 4, 0, 7]
]

# 题目 40 
A40 = [
    [0, 13, 8, 4],
    [4, 9, 8, 4],
    [8, 6, 12, 8],
    [0, 5, 0, -4]
]

# 题目 41 
A41 = [
    [11, -6, 4, -10, -4],
    [-3, 5, -2, 4, 1],
    [-8, 12, -3, 12, 4],
    [1, 6, -2, 3, -1],
    [8, -18, 8, -14, -1]
]

# 执行计算
method_5_1_diagonalize(sp.Matrix(A39), "39")
method_5_1_diagonalize(sp.Matrix(A40), "40")
method_5_1_diagonalize(sp.Matrix(A41), "41")
```


## 5.4 Eigenvectors and Linear Transformations（线性变换的特征向量）

引入更一般的对于特征向量和特征值的定义。

### 线性变换的特征值与特征向量

> **定义**：$V$ 是向量空间，$T:V\to V$ 是线性变换。$T$ 的**特征向量**是非零 $x\in V$，使 $T(x)=\lambda x$；$\lambda$ 称为**特征值**。

例：左双移位 $D(\{x_k\})=\{x_{k+2}\}$ 作用在信号 $\{s_k\}=\{\cos(k\pi/2)\}$ 上：$s_{k+2}=\cos((k+2)\pi/2)=\cos(k\pi/2+\pi)=-\cos(k\pi/2)=-s_k$，故 $D(\{s_k\})=-\{s_k\}$，该信号是特征值 $-1$ 的特征向量。

### 变换的矩阵表示（B-矩阵）

设 $V$ 是 $n$ 维空间，$\mathcal{B}=\{b_1,\dots,b_n\}$ 是一组基。$T:V\to V$ 关于 $\mathcal{B}$ 的**矩阵表示**为
$$
[T]_{\mathcal{B}}=\big[\,[T(b_1)]_{\mathcal{B}}\;\; [T(b_2)]_{\mathcal{B}}\;\cdots\; [T(b_n)]_{\mathcal{B}}\,\big]
$$
它满足 $[T(x)]_{\mathcal{B}}=[T]_{\mathcal{B}}\,[x]_{\mathcal{B}}$。

### 对角矩阵表示（定理 8）

> **定理8**：若 $A=PDP^{-1}$（$D$ 对角），且 $\mathcal{B}$ 是 $P$ 的列构成的 $\mathbb{R}^n$ 的基，则 $D$ 是变换 $x\mapsto Ax$ 的 B-矩阵，即 $[T]_{\mathcal{B}}=D$。

证明：$[T]_{\mathcal{B}}=P^{-1}AP$，而 $A=PDP^{-1}\Rightarrow P^{-1}AP=D$。

#### 矩阵表示的相似性

相似的本质：同一个线性变换在不同基下的表示矩阵是相似的。

更一般地，**$A$ 相似于 $C$（$A=PCP^{-1}$）当且仅当 $C$ 是 $x\mapsto Ax$ 关于以 $P$ 的列为基的 B-矩阵**。

>所有与 $A$ 相似的矩阵集合，恰好就是 $x\mapsto Ax$ 的全部矩阵表示。

数值技巧：算 $P^{-1}AP$ 可直接对增广阵 $[P\mid AP]$ 行化简成 $[I\mid P^{-1}AP]$，不必单独求 $P^{-1}$。

## 5.5 Complex Eigenvalues（复特征值）

### 为什么要研究复数

实矩阵 $A$ 的特征方程 $\det(A-\lambda I)=0$ 是 $n$ 次多项式，在复数域内**恰有 $n$ 个根（计重数）**。若实矩阵 $A$ 有复根，这些复根蕴含着 $A$ 的"旋转"信息（振动、周期运动、空间旋转等现实问题）。

当实矩阵出现复特征值时，意味着什么变换？   **旋转与缩放**

$C^n$ 上的矩阵特征值/特征向量理论完全照搬 $\mathbb{R}^n$ 情形。

### 实矩阵的复特征值成对出现（共轭配对定理）

> 设 $A$ 是实矩阵，$\lambda$ 是复特征值，$x$ 是 $C^n$ 中的对应特征向量。对**实**矩阵有 $\overline{Ax}=A\overline{x}$，于是 $A\overline{x}=\overline{\lambda}\,\overline{x}$。所以 $\overline{\lambda}$ 也是 $A$ 的特征值，$\overline{x}$ 是对应特征向量。
>
> **结论：实矩阵的复特征值（及复特征向量）成共轭对出现。**

### 复向量的实部与虚部

$x=\operatorname{Re}x+i\,\operatorname{Im}x$，其中 $\operatorname{Re}x,\operatorname{Im}x\in\mathbb{R}^n$。

### 基本构件（Example 6）

设 $C=\begin{bmatrix}a&-b\\b&a\end{bmatrix}$（$a,b$ 实，不全为 0）。特征值为 $a\pm bi$；若 $r=|\lambda|=\sqrt{a^2+b^2}$，则
$$
C=r\begin{bmatrix}\cos\varphi&-\sin\varphi\\\sin\varphi&\cos\varphi\end{bmatrix}
$$
即 $x\mapsto Cx$ 是"旋转角度 $\varphi$ + 缩放因子 $r$"的复合（$\varphi$ 是 $\lambda$ 的辐角）。对应特征向量为 $(1,i)$、$(1,-i)$。

### 定理 9（复特征值 → 实旋转矩阵）

> 设 $A$ 是实 $2\times2$ 矩阵，有复特征值 $\lambda=a-bi\;(b\neq0)$ 及 $C^2$ 中特征向量 $v$。则
> $$
> A=PCP^{-1},\qquad P=[\operatorname{Re}v\;\;\operatorname{Im}v],\qquad C=\begin{bmatrix}a&-b\\b&a\end{bmatrix}
> $$



代数推导：

已知$ A \bm{v} = \lambda \bm{v}$ ,其中 $ \lambda = a-bi$ , $ \bm{v} = \bm{u} + i \bm{w} $

代入方程得到 $ A (\bm{u}+i \bm{w}) = (a-bi)(\bm{u}+ i \bm{w})$

展开后实部虚部分别相等，得到
$$
实部：A \bm{u} = a \bm{u} +b \bm{w} \newline
虚部：A \bm{w} = -b\bm{u} + a \bm{w}
$$

设$ P =[\bm{u}\ \bm{w}] $ 

$AP = A [\bm{u}\  \bm{w}] = [a \bm{u} +b \bm{w}\  -b\bm{u} + a \bm{w}]$

$PC = [\bm{u}\  \bm{w}] \begin{bmatrix}a&-b\\b&a\end{bmatrix} =  [a \bm{u} +b \bm{w}\  -b\bm{u} + a \bm{w}]$

即有$AP = PC$ ，且矩阵P可逆
则
$$
A=PCP^{-1}
$$

>任何带有复数特征值的$ 2 \times 2$实矩阵A，本质上就是一个旋转+缩放矩阵C，只是被换了个坐标系。在实际工程计算中，我们不想处理复数矩阵，定理9让我们只用实数矩阵P和C就能精确描述系统。


### 为什么复特征向量的实部和虚部线性无关？

>如果 $\lambda$ 是**真正**的复数（即 $\lambda = a - bi, b \neq 0$），那么对应的实部向量和虚部向量必定线性无关。

设 $A$ 是一个**实矩阵**（$A \in \mathbb{R}^{n \times n}$），$\lambda = a - bi$ 是 $A$ 的一个复数特征值，其中 $a, b \in \mathbb{R}$ 且 $b \neq 0$。
对应的特征向量 $\mathbf{x} \in \mathbb{C}^n$ 满足：
$$A\mathbf{x} = \lambda\mathbf{x}$$

将特征向量 $\mathbf{x}$ 拆分为实部和虚部：
$$\mathbf{x} = \mathbf{u} + i\mathbf{w}$$
其中 $\mathbf{u} = \text{Re}(\mathbf{x}), \mathbf{w} = \text{Im}(\mathbf{x})$。

假设 $\mathbf{u}$ 和 $\mathbf{w}$ 在 $\mathbb{R}^n$ 中**线性相关**。
因为特征向量 $\mathbf{x} \neq \mathbf{0}$，所以 $\mathbf{u}$ 和 $\mathbf{w}$ 不可能同时为零向量。
根据线性相关的定义，存在一个实数 $k$，使得其中一个向量是另一个的实数倍。假设 $\mathbf{w} \neq \mathbf{0}$，则存在 $k \in \mathbb{R}$ 使得：
$$\mathbf{u} = k\mathbf{w}$$

那么，特征向量可以写成：
$$\mathbf{x} = \mathbf{u} + i\mathbf{w} = k\mathbf{w} + i\mathbf{w} = (k + i)\mathbf{w}$$

将 $\mathbf{x} = (k + i)\mathbf{w}$ 代入 $A\mathbf{x} = \lambda\mathbf{x}$ 中：
$$A((k + i)\mathbf{w}) = (a - bi)((k + i)\mathbf{w})$$

因为 $A$ 是实矩阵，标量 $(k + i)$ 可以提出到矩阵乘法外面：
$$(k + i)A\mathbf{w} = (a - bi)(k + i)\mathbf{w}$$

由于 $(k + i) \neq 0$，我们可以同时约掉它：
$$A\mathbf{w} = (a - bi)\mathbf{w}$$

观察上面这个等式：
* **左边** $A\mathbf{w}$：$A$ 是实矩阵，$\mathbf{w}$ 是实向量，所以 $A\mathbf{w}$ 必定是一个**实向量**（每个分量都是实数）。
* **右边** $(a - bi)\mathbf{w}$：因为 $b \neq 0$，$(a - bi)$ 是一个复数。一个复数乘以一个实向量，结果必定包含**虚部**。

一个实向量不可能等于一个含有虚部的向量，除非虚部为 $0$。但我们的前提是 $b \neq 0$。
**矛盾** 

这证明了我们的反证假设“$\mathbf{u}$ 和 $\mathbf{w}$ 线性相关”是错误的。

因此，$\text{Re}(\mathbf{x})$ 和 $\text{Im}(\mathbf{x})$ 必须在 $\mathbb{R}^n$ 中**线性无关**。这就保证了矩阵 $P = [\text{Re}(\mathbf{x}) \quad \text{Im}(\mathbf{x})]$ 是可逆的。




高阶情形：$3\times3$ 实矩阵若有复特征值，则在 $\mathbb{R}^3$ 中存在一个平面，$A$ 在其上表现为旋转（可复合缩放），该平面在 $A$ 下不变（invariant）。（任何$3\times3$ 实矩阵至少有一个实特征值）

在二维空间中，一个复特征值意味着整个二维平面都在旋转。

在三维空间中，矩阵A不可能让整个三维空间都旋转（因为三维空间还包含一个实特征值的方向）。所以，A会在三维空间中“切出”一个特殊的二维平面。

在这个特定平面内，A的行为完全等同于我们前面学的二维旋转（外加缩放）。

“不变（invariant）”的意思：如果你拿一个在这个平面内的向量，乘上矩阵 A 之后，它绝对不会跑到平面外面去。它依然留在这个平面内，只是在这个平面里被旋转了。



>求复特征向量技巧：因 $\lambda$ 是特征值，$(A-\lambda I)x=0$ 两行线性相关，只需用**其中一个方程**把一变量用另一变量表示，再适当取整数值。





### 作业 5.5

**练习5.5 27**：设 $A$ 是 $n\times n$ 实矩阵且 $A^T=A$，$x\in C^n$，$q=x^TAx$。下列等号表明 $q$ 是实数（$\bar q=q$）。给每一步理由。
$$
\overline{q}=
\overline{\overline{x^T}Ax} \overset{(a)}{=}
x^T \overline{Ax}\overset{(b)}{=}
x^T A\overline{x}\overset{(c)}{=}
(x^TA\overline{x})^T\overset{(d)}{=}
\overline{x}^TA^T x\overset{(e)}{=}q
$$

**解**：
## 等式与推导理由


#### 步骤 (a)
**推导**：$\overline{\overline{\mathbf{x}}^T A \mathbf{x}} = \mathbf{x}^T \overline{A} \overline{\mathbf{x}}$

**理由**：**复数共轭与矩阵乘积的运算性质。**

对于矩阵乘积的共轭，有 $\overline{UV} = \overline{U}\  \overline{V}$，并且转置的共轭等于共轭的转置，即 $\overline{U^T} = (\overline{U})^T$。
因此，$\overline{\overline{\mathbf{x}}^T A \mathbf{x}} = \overline{(\overline{\mathbf{x}})^T} \overline{A\mathbf{x}} = \mathbf{x}^T \overline{A\mathbf{x}}$。

#### 步骤 (b)
**推导**：$\mathbf{x}^T \overline{A} \overline{\mathbf{x}} = \mathbf{x}^T A \overline{\mathbf{x}}$

**理由**：**矩阵 $A$ 是实矩阵。**

题目已知 $A$ 是实矩阵（real matrix），因此它的共轭等于它自身，即 $\overline{A} = A$。代入后即可消除 $\overline{A}$ 上的共轭符号。

#### 步骤 (c)
**推导**：$\mathbf{x}^T A \overline{\mathbf{x}} = (\mathbf{x}^T A \overline{\mathbf{x}})^T$

**理由**：**标量的转置等于其自身。**

$\mathbf{x}^T$ 是 $1 \times n$ 矩阵，$A$ 是 $n \times n$ 矩阵，$\overline{\mathbf{x}}$ 是 $n \times 1$ 矩阵。它们的乘积是一个 $1 \times 1$ 的矩阵（即一个标量）。一个标量的转置就是它本身。

#### 步骤 (d)
**推导**：$(\mathbf{x}^T A \overline{\mathbf{x}})^T = \overline{\mathbf{x}}^T A^T \mathbf{x}$

**理由**：**矩阵乘积的转置法则（Reverse Order Law）。**

对于矩阵乘积的转置，有 $(ABC)^T = C^T B^T A^T$。
在这里，$A$ 对应 $B$，$\mathbf{x}^T$ 对应 $A$，$\overline{\mathbf{x}}$ 对应 $C$。
因此 $(\mathbf{x}^T A \overline{\mathbf{x}})^T = \overline{\mathbf{x}}^T A^T (\mathbf{x}^T)^T = \overline{\mathbf{x}}^T A^T \mathbf{x}$。

#### 步骤 (e)
**推导**：$\overline{\mathbf{x}}^T A^T \mathbf{x} = q$

**理由**：**题目给定条件 $A^T = A$ 以及 $q$ 的定义。**

因为 $A^T = A$，所以 $\overline{\mathbf{x}}^T A^T \mathbf{x} = \overline{\mathbf{x}}^T A \mathbf{x}$。再根据题目开头给出的 $q$ 的定义 $q = \overline{\mathbf{x}}^T A \mathbf{x}$，即可得到最终结果 $q$。

再求实特征向量：$Ax=\lambda x$ 取实部，因 $A$、$\lambda$ 实：$A(\operatorname{Re}x)=\lambda(\operatorname{Re}x)$。若 $\operatorname{Re}x=0$ 则 $x=i\operatorname{Im}x$，代入得 $A(\operatorname{Im}x)=\lambda(\operatorname{Im}x)$（因 $\lambda$ 实，$i$ 可消去），此时取 $\operatorname{Im}x$。总之 $\operatorname{Re}x$（或 $\operatorname{Im}x$）是 $A$ 的实特征向量。

**实对称矩阵的特征值必定是实数，其特征向量可以取为实向量**



**练习5.5 31**：把 $A$ 分解为 $A=PCP^{-1}$，其中 $C$ 是含形如 $\begin{bmatrix}a&-b\\b&a\end{bmatrix}$ 的 $2\times2$ 块的**分块对角**矩阵（对每个共轭复特征值对，用 $C^4$ 中一个特征向量的实部、虚部作 $P$ 的两列）。


```python
import numpy as np
from numpy import linalg as la

A = np.array([[.7,-1.1,2.0,1.7],
              [-2.0,-4.0,-8.6,-7.4],
              [0,-.5,-1.0,-1.0],
              [1.0,2.8,6.0,5.3]], float)

w, v = la.eig(A)
print("特征值:", np.round(w, 6))

P_cols = []
C_blocks = []
used = np.zeros(len(w), dtype=bool)

for i in range(len(w)):
    if used[i]:
        continue
        
    lam = w[i]
    if abs(lam.imag) > 1e-9:          # 复数特征值
        vi = v[:, i]
        a, b = lam.real, lam.imag
        
        P_cols.append(vi.real)
        P_cols.append(vi.imag)
        
        C_blocks.append(np.array([[a, b], [-b, a]]))
        
        for j in range(len(w)):
            if j != i and abs(w[j] - np.conj(lam)) < 1e-9:
                used[j] = True
                break
    else:                             # 实数特征值
        P_cols.append(v[:, i].real)
        C_blocks.append(np.array([[lam.real]]))

P = np.column_stack(P_cols)

n = len(w) # 矩阵维度 
C = np.zeros((n, n)) # 初始化全零矩阵
current_idx = 0

for block in C_blocks:
    size = block.shape[0] 
    # 将块填入对角线
    C[current_idx:current_idx+size, current_idx:current_idx+size] = block
    current_idx += size

print("\nP =\n", np.round(P, 6))
print("\nC =\n", np.round(C, 6))
print("\n验证 AP == PC:", np.allclose(A@P, P@C))
```


## 5.6 Discrete Dynamical Systems（离散动力系统）

### 核心方法

差分方程 $x_{k+1}=Ax_k$。若 $A$ 可对角化，特征向量 $v_1,\dots,v_n$ 构成基，$|\lambda_1|\ge|\lambda_2|\ge\cdots\ge|\lambda_n|$。把初始向量分解 $x_0=c_1v_1+\cdots+c_nv_n$，则
$$
x_k=c_1\lambda_1^kv_1+\cdots+c_n\lambda_n^kv_n
$$

**长期行为（$k\to\infty$）**：
- 若 $c_1\neq0$ 且 $|\lambda_1|>1$：$x_k\approx c_1\lambda_1^kv_1$，系统以因子 $\lambda_1$ 增长（增长率由主导特征值决定），$x_k$ 的方向趋于 $v_1$（**主导特征向量**）。
- $x_{k+1}\approx\lambda_1 x_k$（$k$ 充分大）。

**主特征值 $\lambda_1$ 决定了系统的最终增长率，而对应的主特征向量 $ v_1$ 决定了系统各分量最终的稳态比例。**

### Graphical Description of Solutions(解的图形描述)

当A是 $ 2\times 2$ 时，可以把 $ x_{k+1} = A x_k$ 视为初始点 $ x_0$ 被反复映射，轨迹图完全由特征值决定。

**$2\times2$ 的几何分类（原点 0）**：
- 两特征值模都 $<1$：0 是**吸引子（attractor）**，所有轨迹趋 0，最大吸引方向沿模较小特征值的特征向量。
- 两特征值模都 $>1$：0 是**排斥子（repeller）**，轨迹远离 0，最大排斥方向沿模较大特征值的特征向量。
- 一个模 $>1$、一个模 $<1$：0 是**鞍点（saddle point）**，吸引方向沿模较小特征向量，排斥方向沿模较大特征向量。



### Change of Variable

解耦：就是把一个“互相牵扯、互相影响”的复杂方程组，变成多个“各自为政、互不干涉”的独立方程。

也是一场视角的切换：

在标准坐标系$ x_1, x_2$下，系统是耦合的，当切换到特征向量坐标系时，沿着特征向量的方向，变换A的作用仅仅是对向量进行伸缩。因此，再特征向量坐标系下，每个分量$y_i$ 都是独立进化的。

**换元解耦**：令 $y_k=P^{-1}x_k$（$P=[v_1\cdots v_n]$），则 $y_{k+1}=Dy_k$，系统被解耦成 $n$ 个独立方程 $y_i(k+1)=\lambda_i y_i(k)$。$y_k$ 就是 $x_k$ 关于特征向量基的坐标。

**用特征向量作为桥梁，把复杂的耦合问题，降维打击成简单的独立问题，最后再变回原貌。**



**复特征值**：模 $>1$ 螺旋外扩（排斥），模 $<1$ 螺旋内缩（吸引），模 $=1$ 沿椭圆轨道。

**例**：阶段矩阵 $A=\begin{bmatrix}0&0&.33\\.18&0&0\\0&.71&.94\end{bmatrix}$ 特征值 $\approx .98,\;-.02\pm.21i$，均模 $<1$ → 种群灭绝。若幼鸟寻巢存活率提高到 50%（.18→.30），主导特征值 $\approx1.01>1$，种群以 1.01 增长。

## 5.7 Applications to Differential Equations（微分方程的应用）

### 核心方法

线性微分方程组 $x'(t)=Ax(t)$（$x$ 是 $t$ 的向量值函数）。解集是 $n$ 维函数空间。

**试探解（特征函数）**：$x(t)=ve^{\lambda t}$。则 $x'(t)=\lambda ve^{\lambda t}$，$Ax(t)=Ave^{\lambda t}$。故
$$
x'=Ax \iff \lambda v=Av \iff \lambda\text{ 是 }A\text{ 的特征值，}v\text{ 是对应特征向量}
$$
即每个特征值-特征向量对给出一个解 $x(t)=ve^{\lambda t}$（称为**特征函数**）。


#### Decoupling a Dynamical System（解耦一个动力系统）
**可对角化情形（解耦）**：$A=PDP^{-1}$，令 $y(t)=P^{-1}x(t)$，则 $y'=Dy$，解耦为 $y_i'=\lambda_iy_i$，得 $y_i(t)=c_ie^{\lambda_it}$。于是
$$
x(t)=Py(t)=c_1v_1e^{\lambda_1t}+\cdots+c_nv_ne^{\lambda_nt}
$$
初值 $x(0)=x_0$ 确定 $c_i$（解 $Pc=x_0$）。


### Complex Eigenvalues （**复特征值**）


现实世界的物理系统（如电路、弹簧）都是实数，但描述它们的矩阵 A 却常常产生复数特征值。我们如何从复数解中，提取出有物理意义的实数解？


####  复解的出现
假设 $A$ 是实矩阵，它有一对共轭复特征值 $\lambda = a + bi$ 和 $\overline{\lambda} = a - bi$（其中 $b \neq 0$），对应的特征向量为 $\mathbf{v}$ 和 $\overline{\mathbf{v}}$。
根据前面的理论，我们可以直接写出两个复数形式的解：
$$\mathbf{x}_1(t) = \mathbf{v}e^{\lambda t} \quad \text{和} \quad \mathbf{x}_2(t) = \overline{\mathbf{v}}e^{\overline{\lambda} t} \quad (6)$$

####  为什么不能直接用复解？
虽然在电气工程中复数很好用，但物理世界（电压、位移、温度）都是实数。我们无法在现实中测量出一个“复数电压”。因此，必须找到两个**实数形式的解**。

####  欧拉公式（Euler's Formula）的引入
为了解决这个问题，引入了微积分中的欧拉公式。通过指数函数的幂级数展开，可以得到：
$$e^{(a+bi)t} = e^{at}e^{ibt} = e^{at}(\cos bt + i\sin bt) \quad (7)$$
这完美地将复指数拆解成了实部（$e^{at}\cos bt$）和虚部（$e^{at}\sin bt$）。

####  提取实解
将 $\mathbf{v}$ 拆成实部和虚部：$\mathbf{v} = \text{Re}(\mathbf{v}) + i\text{Im}(\mathbf{v})$。
代入 $\mathbf{x}_1(t) = \mathbf{v}e^{\lambda t}$，展开：
$$\mathbf{v}e^{\lambda t} = (\text{Re}(\mathbf{v}) + i\text{Im}(\mathbf{v}))(e^{at})(\cos bt + i\sin bt)$$
利用复数乘法展开并整理，提取出实部和虚部：
*   **实部（Real Part）**：$[(\text{Re}\mathbf{v})\cos bt - (\text{Im}\mathbf{v})\sin bt]e^{at}$
*   **虚部（Imaginary Part）**：$[(\text{Re}\mathbf{v})\sin bt + (\text{Im}\mathbf{v})\cos bt]e^{at}$

**因为原矩阵 $A$ 是实矩阵，所以这个复数解的实部和虚部，各自都是原微分方程 $\mathbf{x}' = A\mathbf{x}$ 的实数解。**
于是得到了两个线性无关的实解 $\mathbf{y}_1(t)$ 和 $\mathbf{y}_2(t)$（当 $b \neq 0$ 时）。

>$e^{(a+bi)t}=e^{at}(\cos bt+i\sin bt)$。取复解 $x_1(t)=ve^{\lambda t}$ 的实部与虚部得到两个实线性无关解：
>$$
y_1(t)=\big[\operatorname{Re}v\cos bt-\operatorname{Im}v\sin bt\big]e^{at},\qquad
y_2(t)=\big[\operatorname{Re}v\sin bt+\operatorname{Im}v\cos bt\big]e^{at}
$$



**几何分类（$2\times2$，$t\to\infty$）**：
- 特征值均为负实：0 是**吸引子（sink）**，最大吸引方向沿"更负"特征值对应特征向量。
- 特征值均为正实：0 是**排斥子（source）**。
- 一正一负实：0 是**鞍点（saddle point）**（先趋近再远离）。
- 复特征值（实部 $<0$ 内旋、$>0$ 外旋、$=0$ 椭圆）：0 是**螺线点（spiral point）**。

与 5.6 区别：5.6 离散系统看 $|\lambda|<1$ 收敛；这里连续系统看**实部** $\operatorname{Re}\lambda<0$ 使 $e^{\lambda t}\to0$。


## 5.8 Iterative Estimates for Eigenvalues（特征值的迭代估计）

### **为什么特征值难以精确求解？**
#### 1. 数学理论壁垒（代数局限）
*   **五次及以上无根式解**：求特征值等价于求特征多项式 $\det(A - \lambda I) = 0$ 的根。
*   根据**阿贝尔-鲁菲尼定理**，对于 $n \ge 5$ 的矩阵（即五次及以上多项式），**不存在通用的代数求根公式**。绝大多数高阶矩阵无法用有限步代数运算表达出精确解。

#### 2. 数值计算灾难（浮点误差与病态）
*   **计算多项式极不稳定**：如果先算出特征多项式系数再用计算机求根，舍入误差会被剧烈放大。
*   **病态条件（Ill-conditioned）**：著名的“威尔金森多项式”表明，多项式系数的微小扰动（如计算机浮点截断误差）会导致根发生巨大变化。特征值极度接近或矩阵条件数大时，精确解毫无现实意义。

#### 3. 工程实际需求（够用即可）
*   在结构力学、PageRank 算法等实际应用中，矩阵维度动辄成千上万。
*   工程师**不需要**绝对精确的符号解（如 $\lambda = 3$），只需要**数值近似解**（如 $3.000001$）。只要误差在可接受范围内即可满足工程需求。




### 幂法（Power Method）

适用于 $A$ 有**严格主导特征值（ strictly dominant eigenvalue ）** $\lambda_1$（$|\lambda_1|>|\lambda_2|\ge\cdots\ge|\lambda_n|$）。若 $x=c_1v_1+\cdots+c_nv_n$ 且 $c_1\neq0$，
$$
A^kx=c_1\lambda_1^kv_1+c_2\lambda_2^kv_2+\cdots+c_n\lambda_n^kv_n
\;\Longrightarrow\;
\frac{1}{\lambda_1^k}A^kx\to c_1v_1
$$
即 $A^kx$ 的方向趋于特征向量 $v_1$。

**幂法算法**：
1. 取最大分量为 1 的初始向量 $x_0$；
2. 对 $k=0,1,\dots$：a. 算 $Ax_k$；b. 令 $\mu_k$ 为 $Ax_k$ 中绝对值最大的分量；c. $x_{k+1}=\frac1{\mu_k}Ax_k$；
3. 则 $\mu_k\to\lambda_1$，$x_k\to$ 对应特征向量（$v_1$ 的某倍数）。

收敛速率由比值 $|\lambda_2/\lambda_1|$ 决定：越接近 1 收敛越慢。

### 反幂法（Inverse Power Method）

估计任一特征值 $\lambda$，需要较好的初估 $\alpha$。令 $B=(A-\alpha I)^{-1}$：$B$ 的特征值为 $\frac1{\lambda_i-\alpha}$，特征向量与 $A$ 相同。若 $\alpha$ 很接近 $\lambda_2$，则 $\frac1{\lambda_2-\alpha}$ 是 $B$ 的严格主导特征值，对 $B$ 用幂法即可快速逼近 $\lambda_2$。

**反幂法算法**：对每步求解 $(A-\alpha I)y_k=x_k$（用 LU 分解加速），取 $y_k$ 最大分量 $\mu_k$，令 $\eta_k=\alpha+\frac1{\mu_k}$、$x_{k+1}=\frac1{\mu_k}y_k$。则 $\eta_k\to\lambda$。

取 $\alpha=0$ 可估计最接近 0 的特征值。

**练习5.8 13**：3 阶矩阵 $A$ 的特征值约为 $4,-4,3$。若接近 4、4 的两个特征值**模不同**，幂法能用吗？是否可能有用？



**答：能。**
**理由**：
幂法能够成功的先决条件是矩阵 $A$ 拥有一个**严格占优特征值（strictly dominant eigenvalue）**。即存在一个特征值 $\lambda_1$，使得其绝对值严格大于其他所有特征值的绝对值：$|\lambda_1| > |\lambda_2| \ge \dots \ge |\lambda_n|$。

题目明确指出，已知接近 4 和 -4 的特征值具有**不同的绝对值**。
*   假设 $\lambda_1 \approx 4.1$， $\lambda_2 \approx -4.0$。最大绝对值是 4.1。
*   或者假设 $\lambda_1 \approx -4.1$， $\lambda_2 \approx 4.0$。最大绝对值是 4.1。
无论哪种情况，都存在一个绝对值的“王者”。因此，从数学原理上讲，幂法**一定会收敛**到那个具有最大绝对值的特征值（尽管这个特征值可能是正数，也可能是负数）。

**但是不太有用**

**理由**：
虽然幂法能起作用（会收敛），但它的**收敛速度极度依赖于比值 $|\lambda_2 / \lambda_1|$**。
*   收敛速度公式：误差项以 $(\lambda_2 / \lambda_1)^k$ 的速率衰减。
*   本题中，$\lambda_1$ 和 $\lambda_2$ 的绝对值非常接近（都约为 4）。比值 $|\lambda_2 / \lambda_1| \approx 1$。
*   当这个比值极其接近 1 时，$(\lambda_2 / \lambda_1)^k$ 衰减得极其缓慢。这意味着幂法将需要成百上千次迭代才能收敛到令人满意的精度。




**练习5.8 17**（反幂法）：估计 Example 3 中 $A=\begin{bmatrix}10&-8&-4\\-8&13&4\\-4&5&4\end{bmatrix}$ 的**中间**特征值，精度 4 位小数，$x_0=(1,0,0)$。

 


```python
from numpy import linalg as la
A = np.array([[10,-8,-4],[-8,13,4],[-4,5,4]], float)
print("精确特征值:", np.round(la.eigvals(A),6))
# 中间特征值约 3.3，用反幂法
alpha = 3.3
x = np.array([[1.0],[0],[0]])
for k in range(1,8):
    y = la.solve(A - alpha*np.eye(3), x)
    mu = y.flatten()[np.argmax(np.abs(y.flatten()))]
    eta = alpha + 1/mu
    x = y/mu
    print(f"k={k}: eta={eta:.6f}, xk={x.flatten()}")
```

**结果**：精确特征值约 $2,\;3.321,\;21.679$，中间特征值 $=3.3212$。取 $\alpha=3.3$ 用反幂法，几步内 $\eta_k\to3.3212$。

```python
x = np.array([.5,.5])
for name, A in [("a", [[.8,0],[0,.2]]), ("b", [[1,0],[0,.8]]), ("c", [[8,0],[0,2]])]:
    A=np.array(A,float); w= np.linalg.eigvals(A)
    print(f"\n{name} 特征值={w}")
    for k in (1,3,10):
        print(f"  A^{k} x = {np.linalg.matrix_power(A,k)@x}")
```

## 5.9 Applications to Markov Chains（马尔可夫链的应用）

### 基本概念

> **定义**：**概率向量（probability vector）** 是各项非负且和为 1 的向量；**随机矩阵（stochastic matrix）** 是各列都是概率向量的方阵。
>
> **马尔可夫链**：概率向量序列 $x_0,x_1,\dots$ 配合随机矩阵 $P$，满足 $x_{k+1}=Px_k$（$k=0,1,\dots$）。$x_k$ 称为**状态向量**，分量 $=$ 系统处于各状态的概率。

### 定理 10 / 11

> **定理10**：若 $P$ 是随机矩阵，则 1 是 $P$ 的特征值。
> **证明**：$P$ 各列和为 1 $\Rightarrow$ $P^T$ 各行和为 1 $\Rightarrow P^T\mathbf e=\mathbf e$（$\mathbf e=(1,\dots,1)^T$），故 1 是 $P^T$ 的特征值；$P$ 与 $P^T$ 特征值相同（5.2 练习 20），故 1 是 $P$ 的特征值。

> **定义**：随机矩阵 $P$ 的**稳态向量（steady-state / equilibrium vector）** 是满足 $Pq=q$ 的概率向量 $q$。稳态向量是特征值 1 的特征向量，再归一化成概率向量（各分量除以其和）。

> **定义**：随机矩阵 $P$ **正则（regular）**，若某次幂 $P^k$ 的每个元素都严格为正。

> **定理11**：若 $P$ 是 $n\times n$ **正则**随机矩阵，则 $P$ 有唯一稳态向量 $q$；且对任意初始状态 $x_0$，马尔可夫链 $x_{k+1}=Px_k$ 收敛到 $q$（$k\to\infty$）。**初始状态不影响长期行为**。

求稳态向量：解 $(P-I)x=0$，取特征值 1 的特征向量 $w$，再归一化 $q=w/(\text{各项和})$。（技巧：给增广矩阵乘 10 化整小数，但不要只给 $P$ 乘 10，要给整个增广矩阵乘。）

另：$x_k=P^kx_0$。

```python

P = np.array([[.70,.15],[.30,.85]])
w,v = la.eig(P); print("特征值:",w)
# 特征值1的特征向量 -> 归一化为概率向量
q = v[:,np.argmin(np.abs(w-1))].real
if q.sum()<0: q=-q
q = q/q.sum()
print("稳态向量 q =", np.round(q,4))
  
```

**练习5.9 27**：设 $P$ 是 $n\times n$ 随机矩阵，证明 $Px=x$ 有非平凡解。逐条论证：

**a.** 把 $P-I$ 的其它各行都加到最底行，结果得到全零行。

**证明**：$P$ 每列和为 1，故 $P-I$ 每列和为 $1-1=0$。把所有行加到最底行时，最底行的第 $j$ 列变为"第 $j$ 列各行之和 $=0$"，故为全零行。

**b.** $P-I$ 的行线性相关。

**证明**：由 (a)，$P-I$ 经行运算产生零行，行化简后有非主元行，故行线性相关（行秩 $<n$）。

**c.** $P-I$ 的行空间维数小于 $n$。

**证明**：行空间维数 = 行秩 $<n$。

**d.** $P-I$ 有非平凡零空间。

**证明**：$\dim(\operatorname{Null}(P-I))=n-\operatorname{rank}(P-I)>0$（秩-零化度定理），故存在非零 $x$ 使 $(P-I)x=0$，即 $Px=x$。结合随机矩阵特征值理论可进一步得到非负的稳态解。




**练习5.9 28**：证明每个 $2\times2$ 随机矩阵至少有一个稳态向量。可写成 $P=\begin{bmatrix}1-\alpha&\beta\\\alpha&1-\beta\end{bmatrix}$（$0\le\alpha,\beta\le1$）。



**证明**：解 $(P-I)q=0$：
$$
P-I=\begin{bmatrix}-\alpha&\beta\\\alpha&-\beta\end{bmatrix}

方程 $-\alpha q_1+\beta q_2=0$，即 $\alpha q_1=\beta q_2$。取 $q_1=\beta,\;q_2=\alpha$，得特征向量 $(\beta,\alpha)$，归一化为 $q=\frac1{\alpha+\beta}(\beta,\alpha)$（当 $\alpha+\beta>0$ 时）。若 $\alpha=\beta=0$，则 $P=I$，任意概率向量都是稳态向量（两个线性无关的稳态向量）。故每个 $2\times2$ 随机矩阵至少有一个稳态向量（$\alpha=\beta=0$ 时有无穷多，否则恰一个）。

**证明**：$Sx$ 就是 $x$ 各项之和。故 $x$ 各项非负且和为 1 $\iff$ $x$ 非负且 $Sx=1$，这正是概率向量定义。

$P,Q$ 是正则随机矩阵，由定理 11 有唯一稳态向量 $q$，且对任意初始向量 $x_0$，$x_k=P^kx_0\to q$。因此 $P^k$ 的每一列（= $P^k e_j$，$e_j$ 是标准基向量，也是概率向量）都收敛到同一个 $q$，这就是为什么 $P^k$ 的列趋向相同且趋于稳态向量。

**一般结论**：正则随机矩阵的幂的各列收敛到其唯一稳态向量。





**练习5.9 32**比较求稳态向量的两种方法——(1) 解 $(P-I)x=0$（如 Example 5）；(2) 计算 $P^k$ 取某大 $k$ 的某一列作近似。用你能允许的最大随机随机矩阵实验，$k=100$ 或更大，比较两者的时间与便捷性。


```python
import time

def compare_methods(n, k):
    # 生成一个 n x n 的随机正则随机矩阵 P
    P = np.random.rand(n, n)
    P = P / P.sum(axis=0) 

    #方法 (1)：解方程求零空间 (精确解)
    print(f"\n--- 方法 (1) ---")
    
    # 构造矩阵 A = P - I
    A = P - np.eye(n)
    
    t1_start = time.time()
    u, s, vh = np.linalg.svd(A)
    q1 = vh[-1, :] 
    # 归一化（使其所有元素之和为 1，满足稳态向量条件）
    q1 = q1 / np.sum(q1)
    t1_end = time.time()
    time1 = t1_end - t1_start
    
    print(f"计算耗时: {time1:.6f} 秒")

    #方法 (2)：计算 P 的 k 次幂 (近似解)
    print(f"\n--- 方法 (2) ---")
    
    t2_start = time.time()
    P_k = np.linalg.matrix_power(P, k)
    q2 = P_k[:, 0]
    t2_end = time.time()
    time2 = t2_end - t2_start
    
    print(f"计算耗时: {time2:.6f} 秒")

    # 打印前5个数值进行对比
    print(f"方法(1) 稳态向量前5个分量: {q1[:5]}")
    print(f"方法(2) 稳态向量前5个分量: {q2[:5]}")
    print(f"两者向量间的最大差异: {np.max(np.abs(q1 - q2)):.2e}")

# 运行实验 (将矩阵设为50阶，幂次设为100)
compare_methods(50, 100)
```



**讨论要点**：方法 (1) 解线性方程组一次即可，精确、快，但需行化简；方法 (2) 反复矩阵乘法，$P^k$ 收敛速度取决于次主导特征值（$\lambda_2$，$\le1$）的大小，若 $|\lambda_2|$ 接近 1 收敛很慢，需要很大 $k$。

**补充 40**：设 $A=\begin{bmatrix}.4&-.3\\.4&1.2\end{bmatrix}$（解释为什么 $A^k$ 当 $k\to\infty$ 时趋于矩阵 $L=\begin{bmatrix}-.5&-.75\\1&1.5\end{bmatrix}$（即每行成比例的秩 1 矩阵）。


**解**：

计算矩阵 $A$ 的特征多项式：
$$ \det(A - \lambda I) = (\lambda - 1)(\lambda - .6) = 0 $$

解得特征值：
*   $\lambda_1 = 1$
*   $\lambda_2 = .6$

因为 $A$ 有两个不同的特征值，所以 $A$ 可以对角化为：
$$ A = PDP^{-1} \implies A^k = P D^k P^{-1} $$

其中 $D^k = \begin{bmatrix} 1^k & 0 \\ 0 & (.6)^k \end{bmatrix}$。
当 $k \to \infty$ 时，由于 $|.6| < 1$，所以 $(.6)^k \to 0$。

因此，极限状态下的对角矩阵为：
$$ \lim_{k \to \infty} D^k = \begin{bmatrix} 1 & 0 \\ 0 & 0 \end{bmatrix} $$


对应的特征向量：
*   对于 $\lambda_1 = 1$，解 $(A-I)x=0$ 得 $\mathbf{p}_1 = \begin{bmatrix} 1 \\ -2 \end{bmatrix}$。
*   对于 $\lambda_2 = .6$，解 $(A-.6I)x=0$ 得 $\mathbf{p}_2 = \begin{bmatrix} -3 \\ 2 \end{bmatrix}$。

构造矩阵 $P$ 及其逆矩阵：
$$ P = \begin{bmatrix} 1 & -3 \\ -2 & 2 \end{bmatrix}, \quad P^{-1} = -\frac{1}{4} \begin{bmatrix} 2 & 3 \\ 2 & 1 \end{bmatrix} = \begin{bmatrix} -.5 & -.75 \\ -.5 & -.25 \end{bmatrix} $$

计算极限矩阵 $\lim_{k \to \infty} A^k$：

$$
\begin{aligned}
\lim_{k \to \infty} A^k &= P \begin{bmatrix} 1 & 0 \\ 0 & 0 \end{bmatrix} P^{-1} \\
&= \begin{bmatrix} 1 & -3 \\ -2 & 2 \end{bmatrix} \begin{bmatrix} 1 & 0 \\ 0 & 0 \end{bmatrix} \begin{bmatrix} -.5 & -.75 \\ -.5 & -.25 \end{bmatrix} \\
&= \begin{bmatrix} 1 & 0 \\ -2 & 0 \end{bmatrix} \begin{bmatrix} -.5 & -.75 \\ -.5 & -.25 \end{bmatrix} \\
&= \begin{bmatrix} -.5 & -.75 \\ 1.0 & 1.50 \end{bmatrix}
\end{aligned}
$$

### 结论
由于 $\lambda_1=1$ 而 $|\lambda_2|<1$，随着迭代次数 $k$ 增加，$\lambda_2$ 对应的部分衰减为 0，最终仅剩 $\lambda_1=1$ 对应的稳态部分，即矩阵 $\begin{bmatrix} -.5 & -.75 \\ 1.0 & 1.50 \end{bmatrix}$。

## 第五章总结

- **特征值求法**：解特征方程 $\det(A-\lambda I)=0$；三角矩阵直接读主对角线。
- **特征向量求法**：对每个 $\lambda$ 行化简 $(A-\lambda I\mid0)$ 求特征空间基。
- **对角化**：$n$ 个线性无关特征向量 $\Leftrightarrow A=PDP^{-1}$；$A^k=PD^kP^{-1}$。
- **相似**：$P^{-1}AP$ 保持特征值、迹、行列式、可对角化性。
- **应用**：
  - 差分方程 $x_{k+1}=Ax_k$：$x_k=\sum c_i\lambda_i^kv_i$（离散，看 $|\lambda|$）；
  - 微分方程 $x'=Ax$：$x(t)=\sum c_iv_ie^{\lambda_it}$（连续，看 $\operatorname{Re}\lambda$）；
  - 马尔可夫链：正则随机矩阵收敛到唯一稳态向量（$\lambda=1$ 的特征向量）。
- **复特征值**：实矩阵成共轭对，对应平面旋转（定理 9），$A=PCP^{-1}$。
- **迭代估计**：幂法（严格主导特征值）、反幂法（任意特征值，需初估）、Rayleigh 商（对称时更高精度）。
